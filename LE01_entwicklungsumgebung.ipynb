## <a id="kap-gpu-und-cuda-was-sie-heute-brauchen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.5&nbsp;&nbsp;GPU und CUDA: Was Sie heute brauchen</span>

**CUDA** ist die Plattform, mit der Programme auf NVIDIA-Grafikkarten rechnen. Das ist der Weg, über den in diesem Kurs später jeder GPU-Code läuft. Eine CUDA-GPU haben Sie also genau dann, wenn in Ihrem Rechner eine NVIDIA-Karte steckt. Alles andere — Macs (auch mit Apple-Chip), AMD- und Intel-Grafik — kann vieles, aber kein CUDA.

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Warum ausgerechnet NVIDIA und was CUDA technisch ist, klärt die GPU-Lerneinheit. Für die Einordnung heute reicht: CUDA ist der Industriestandard. PyTorch unterstützt zwar auch andere Backends (etwa Apple-Chips und AMD-Karten), am besten ausgebaut ist aber der CUDA-Pfad, und die großen KI-Rechenzentren setzen fast durchweg auf NVIDIA. Fürs Training von Modellen ist CUDA damit weiterhin die erste Wahl; bei der Inferenz, also dem Ausführen fertiger Modelle, haben Alternativen wie TPUs, AMD und spezialisierte Chips inzwischen echte Anteile.</p></div></details>

Und jetzt die wichtigste Nachricht: **Für die ersten Lerneinheiten brauchen Sie noch keine GPU.** Python, NumPy, pandas und die ersten PyTorch-Schritte laufen vollständig auf der CPU. Wenn die GPU-Lerneinheiten kommen, gibt es für jede Situation einen Weg:

| Ihre Situation | Ihr GPU-Weg |
|---|---|
| Rechner mit NVIDIA-Karte | lokal möglich, sofern Grafikkarte, Treiber und eine CUDA-fähige PyTorch-Installation zusammenpassen — der Selbsttest unten sagt es Ihnen |
| Mac, AMD/Intel-Grafik, Laptop ohne GPU | Colab mit T4-GPU (kostenlos) — oder der Hochschul-Server, sobald er da ist |
| gar kein eigener Rechner zur Hand | Colab im Browser reicht aus; nur die präzisen GPU-Messungen später im Semester laufen auf dem Hochschul-Server |

## <a id="kap-jupyter-notebooks-bedienen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.6&nbsp;&nbsp;Jupyter-Notebooks bedienen</span>

Alle Kursmaterialien sind **Jupyter-Notebooks**: Dokumente aus Zellen, in denen sich Text und ausführbarer Code abwechseln. Textzellen erklären, Codezellen rechnen — Code, Ergebnis und Erklärung stehen beieinander, deshalb sind Notebooks das Standardwerkzeug der Daten- und KI-Welt.

**Zellen ausführen.** Klicken Sie eine Codezelle an und drücken Sie **Umschalt+Eingabe**: Die Zelle läuft, das Ergebnis erscheint darunter, die Auswahl springt zur nächsten Zelle. **Strg+Eingabe** führt aus, ohne weiterzuspringen. Neben ausgeführten Zellen steht eine Nummer in eckigen Klammern — die Reihenfolge, in der die Zellen gelaufen sind.

```python
# Probieren Sie es: Zelle anklicken, Umschalt+Eingabe
antwort = 6 * 7
print("Es läuft! Die Antwort ist", antwort)
```

**Der Kernel — das Gedächtnis des Notebooks.** Hinter jedem Notebook läuft ein Python-Prozess, der **Kernel**. Er merkt sich alles, was Ihre Zellen definieren: `antwort` aus der Zelle oben existiert jetzt auch für jede andere Zelle. Deshalb zählt die **Ausführungs-Reihenfolge**, nicht die Position im Dokument — wer mittendrin einsteigt, dem fehlen oft die Variablen von weiter oben.

**Zwei Modi.** Ein Klick in eine Zelle bringt Sie in den **Bearbeitungsmodus** (Sie tippen in der Zelle), **Esc** zurück in den **Kommandomodus**, in dem Tasten das Notebook steuern. In **JupyterLab** (lokal und auf dem Hochschul-Server): `a`/`b` fügt eine Zelle oberhalb/unterhalb ein, zweimal `d` löscht die markierte Zelle. **Colab** hat eigene Kürzel: erst **Strg+M** (Mac: **Cmd+M**), dann `a`, `b` oder `d` — oder Sie nutzen einfach die Schaltflächen, die am Zellenrand erscheinen. Und wenn Sie aus Versehen in eine **Textzelle** doppelgeklickt haben und plötzlich deren Quelltext sehen: **Umschalt+Eingabe** stellt die Ansicht wieder her.

<div style="border-radius:6px; overflow:hidden; margin:2px 0; margin-right:-20px;">
<div style="background:#2E75B6; color:#fff; padding:6px 16px; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Der Reset-Knopf</div>
<div style="background:#EAF2FA; color:#1a1a1a; padding:12px 16px; margin:0;">
<p>Wenn sich ein Notebook seltsam verhält — Variablen fehlen, Ergebnisse passen nicht zur Reihenfolge —, gibt es einen zuverlässigen Neuanfang: Menü <em>Laufzeit</em> (Colab) bzw. <em>Kernel</em> (JupyterLab) → <strong>Neu starten und alle ausführen</strong>. Das Notebook läuft dann einmal sauber von oben nach unten durch. Eine Ausnahme ist erwartbar: Bei noch ungelösten Aufgaben schlagen deren Prüfzellen fehl — das ist kein Installationsfehler. Führen Sie in dem Fall nach dem Neustart die Zellen bis zu Ihrer aktuellen Aufgabe aus und arbeiten Sie dort weiter.</p>

**Speichern.** In Colab arbeiten Sie erst dauerhaft, nachdem Sie einmal **„Kopie in Drive speichern"** geklickt haben — danach speichert Colab automatisch. Die Drive-Kopie sichert dabei nur das Notebook selbst: Dateien, die Sie in der Colab-Sitzung hochladen oder erzeugen, liegen im temporären Speicher der Laufzeit und verschwinden mit ihr — laden Sie solche Dateien bei Bedarf herunter oder legen Sie sie in Drive ab. Lokal und auf dem Hochschul-Server speichert JupyterLab regelmäßig selbst; **Strg+S** (Mac: **Cmd+S**) erzwingt es.

**Die Spielregeln der Kurs-Notebooks.** Prüfzellen führen Sie einfach aus — sie melden „Alle Prüfungen bestanden ✓" oder sagen Ihnen, was nicht stimmt. Vorgabe-Zellen sind gegen versehentliches Löschen gesperrt. Und wo `# TODO: Ihre Lösung hier` steht, sind Sie dran.

```python
# Prüfzelle 2: Sind die Kurs-Pakete da? torch darf anfangs fehlen (kommt später im Semester)
import importlib

for package in ["numpy", "pandas", "matplotlib", "torch"]:
    try:
        module = importlib.import_module(package)
        print(f"✓ {package:12s} {getattr(module, '__version__', '?')}")
    except ModuleNotFoundError:
        hints = {"torch": "kommt später im Semester dazu"}
        hint = hints.get(package, "fehlt — bitte melden")
        print(f"— {package:12s} nicht installiert ({hint})")
    except ImportError as fehler:
        print(f"✗ {package:12s} Import fehlgeschlagen: {fehler}")
```
