## <a id="kap-variablen-und-zahlen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.2&nbsp;&nbsp;Variablen und Zahlen</span>

Eine Variable entsteht durch Zuweisung — ohne Typangabe, ohne Semikolon. Python kennt ganze Zahlen (`int`) und Kommazahlen (`float`) und rechnet mit den gewohnten Operatoren; `**` ist die Potenz, `//` teilt ganzzahlig, `%` liefert den Rest:

```python
students = 25
groups = 4
print(students / groups)     # echte Division -> 6.25 (float)
print(students // groups)    # ganzzahlig    -> 6
print(students % groups)     # Rest          -> 1
print(2 ** 10)               # Potenz        -> 1024
```

## <a id="kap-strings"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.3&nbsp;&nbsp;Strings</span>

Text steht in Anführungszeichen (einfach oder doppelt, egal). Am wichtigsten für diesen Kurs sind **f-Strings**: ein `f` vor dem String, und in geschweiften Klammern dürfen Variablen und Ausdrücke stehen — mit Formatangabe hinter dem Doppelpunkt:

## <a id="kap-listen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.4&nbsp;&nbsp;Listen</span>

Eine Liste sammelt Werte in eckigen Klammern. Sie hat eine Länge (`len`), man greift mit `liste[0]` auf Elemente zu (Zählung ab null!), hängt mit `append` an und prüft mit `in`, ob etwas enthalten ist:

```python
scores = [0.83, 0.41, 0.77]
print(len(scores))       # 3
print(scores[0])         # das erste Element: 0.83
scores.append(0.12)      # hinten anhängen
print(scores)
print(0.41 in scores)    # True
```

## <a id="kap-schleifen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.6&nbsp;&nbsp;Schleifen</span>

Die `for`-Schleife läuft direkt über die Elemente einer Liste — kein Index nötig. Wenn Sie Zahlen von 0 bis n brauchen, gibt `range(n)` sie Ihnen; `while` wiederholt, solange eine Bedingung gilt:

```python
scores = [0.83, 0.41, 0.77, 0.12]

total = 0
for s in scores:          # s nimmt nacheinander jeden Wert an
    total = total + s
print("Summe:", total)

for i in range(3):        # 0, 1, 2
    print("Durchlauf", i)
```

## <a id="kap-funktionen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.7&nbsp;&nbsp;Funktionen</span>

Mit `def` verpacken Sie Logik zum Wiederverwenden. Parameter können Standardwerte haben, `return` gibt das Ergebnis zurück — und eine Funktion, die nichts zurückgibt, liefert automatisch `None`:

```python
def to_percent(score, digits=1):
    return f"{score * 100:.{digits}f} %"

print(to_percent(0.895))       # nutzt den Standardwert digits=1
print(to_percent(0.895, 0))    # überschreibt ihn
```
