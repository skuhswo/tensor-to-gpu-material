Ein Notebook besteht aus Zellen. Codezellen führen Sie mit **Umschalt+Eingabe** aus; das Ergebnis erscheint darunter. Probieren Sie es:

```python
# Alles hinter einem Doppelkreuz ist ein Kommentar — für Menschen, nicht für Python
print("Hallo Tensor to GPU")
print("drei plus vier ist", 3 + 4)   # print nimmt mehrere Werte, Kommas werden zu Leerzeichen
3 + 4
```

Drei Dinge sind hier passiert: Kommentare beginnen mit `#` und werden ignoriert (so sind auch alle Hinweise in den Kurszellen geschrieben), `print` gibt beliebig viele Werte durch Kommas getrennt aus, und der **letzte** Ausdruck einer Zelle wird zusätzlich automatisch angezeigt — deshalb sehen Sie die 7 ohne `print`.

Drei Handgriffe, die Sie vor Frust bewahren:

1. **Reihenfolge zählt, nicht Position.** Zellen wirken in der Reihenfolge, in der Sie sie ausführen. Wenn sich das Notebook seltsam verhält: Kernel neu starten (Menü *Laufzeit* bzw. *Kernel*) und die Zellen von oben bis zu Ihrer aktuellen Aufgabe erneut ausführen. *Neu starten und alle ausführen* ist die Schlusskontrolle, wenn alle Aufgaben gelöst sind — vorher scheitern die Prüfzellen der noch offenen Aufgaben erwartungsgemäß.
2. **Fehlermeldungen von unten lesen.** Die letzte Zeile des Tracebacks nennt Fehlertyp und Problem, direkt darüber steht die Zeile Ihres Codes, in der es passiert ist.
3. **Aus Versehen in einer Textzelle gelandet?** Umschalt+Eingabe stellt die Ansicht wieder her.

**Und wenn es knallt?** Dann lesen Sie den Traceback, finden die Zeile, korrigieren und führen erneut aus — dieser Zyklus ist das halbe Programmieren, und Sie üben ihn gleich einmal. Zwei Fehlertypen begegnen Ihnen dabei: Ein `TypeError` heißt, die Typen passen nicht zusammen (etwa eine Zahl minus einen String), ein `NameError` heißt, diesen Namen gibt es nicht — meist ein Tippfehler.

## <a id="kap-variablen-und-zahlen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.3&nbsp;&nbsp;Variablen und Zahlen</span>

Eine Variable entsteht durch Zuweisung — ohne Typangabe, ohne Semikolon. Der Typ hängt dabei am **Wert**, nicht an der Variablen: `type()` verrät ihn, und derselbe Name darf nacheinander Werte verschiedener Typen tragen (in C++ undenkbar). Python kennt ganze Zahlen (`int`, beliebig groß) und Fließkommazahlen (`float`) und rechnet mit den gewohnten Operatoren; `**` ist die Potenz, `//` teilt ganzzahlig, `%` liefert den Rest:

```python
students = 25
groups = 4
print(students / groups)     # echte Division -> 6.25 (float)
print(students // groups)    # ganzzahlig    -> 6
print(students % groups)     # Rest          -> 1
print(2 ** 100)              # int wird beliebig groß — kein Überlauf
print(type(students), type(students / groups))
```

Merken Sie sich den Unterschied der Divisionen: `/` liefert **immer** ein `float`, auch bei 8 / 2. Wer ein ganzzahliges Ergebnis braucht (etwa als Index), nimmt `//` — mit zwei Feinheiten: `//` rundet nach **unten**, bei negativen Zahlen also anders als die C/C++-Ganzzahldivision (`-7 // 3` ist `-3`, nicht `-2`), und ein `int` kommt nur heraus, wenn beide Operanden `int` sind (`8.0 // 2` ergibt `4.0`).

Zwischen den Typen wandeln Sie ausdrücklich um — die Typnamen sind zugleich Funktionen:

```python
print(int("42") + 1)             # String -> Zahl
print(float("0.5") * 2)
print("Ergebnis: " + str(7))     # Zahl -> String, zum Verketten
```

## <a id="kap-strings"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.4&nbsp;&nbsp;Strings</span>

Text steht in Anführungszeichen (einfach oder doppelt, egal); ein eigenes `char` gibt es nicht, und Strings sind unveränderlich. Am wichtigsten für diesen Kurs sind **f-Strings**: ein `f` vor dem String, und in geschweiften Klammern dürfen Variablen und Ausdrücke stehen — mit Formatangabe hinter dem Doppelpunkt:

```python
name = "Ada"
score = 0.83124
print("Hallo " + name + "!")                   # + verkettet Strings
print(f"{name} erreicht {score} Punkte")
print(f"{name} erreicht {score:.2f} Punkte")   # :.2f = zwei Nachkommastellen
print(f"in Prozent: {score * 100:.0f} %")      # auch Ausdrücke sind erlaubt
```

## <a id="kap-listen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.5&nbsp;&nbsp;Listen</span>

Eine Liste sammelt Werte in eckigen Klammern — wie ein `vector`, nur ohne Typparameter. Sie hat eine Länge (`len`), man greift mit `liste[0]` auf Elemente zu (Zählung ab null!), hängt mit `append` an, entfernt mit `remove` und prüft mit `in`, ob etwas enthalten ist:

```python
scores = [0.83, 0.41, 0.77]
print(len(scores))       # 3
print(scores[0])         # das erste Element: 0.83
scores.append(0.12)      # hinten anhängen
print(scores)
print(0.41 in scores)    # True
scores.remove(0.41)      # entfernt das erste Vorkommen dieses Werts
print(scores)
```

Drei Griffe dazu, die Sie ständig brauchen: Ein **Slice** `scores[:2]` schneidet einen Ausschnitt heraus (vertieft in Teil 2), negative Indizes zählen von hinten, und die eingebauten Funktionen `max`, `min` und `sum` arbeiten direkt auf der ganzen Liste:

```python
scores = [0.83, 0.77, 0.12]
print(scores[:2])        # Slice: die ersten zwei Elemente, als neue Liste
print(scores[-1])        # negative Indizes zählen von hinten: das letzte Element
print(f"{max(scores)}  {min(scores)}  Summe {sum(scores):.2f}")
```

Eine Eigenheit, die regelmäßig Code zerstört — die **`None`-Falle**: Methoden wie `append` und `remove` **verändern die Liste** und geben als Ergebnis `None` zurück. Wer dieses Ergebnis trotzdem zuweist, überschreibt seine Liste: Nach `scores = scores.append(0.9)` ist `scores` keine Liste mehr, sondern `None`, und der nächste Listen-Zugriff stürzt ab. Diese Methoden werden für ihre Wirkung aufgerufen, nicht für einen Rückgabewert.

## <a id="kap-schleifen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.7&nbsp;&nbsp;Schleifen</span>

Das Range-for aus C++ kennen Sie — Pythons `for` ist genau das, nur ist es hier die **einzige** Standardform: Die Schleife läuft direkt über die Elemente, ganz ohne Index. Wenn Sie Zahlen brauchen, gibt `range(n)` sie Ihnen, von 0 bis **ausschließlich** n; mit `range(start, stop)` geben Sie den Startwert vor — `start` gehört dazu, `stop` bleibt ausgeschlossen. Und wenn Sie beim Durchlaufen doch die Position brauchen, liefert `enumerate` je Durchlauf ein Paar aus Nummer und Element — die zwei Namen vor dem `in` übernehmen dessen Bestandteile (mehr zu Paaren im Kapitel <a href="#kap-funktionen">1.8</a>):

```python
scores = [0.83, 0.41, 0.77, 0.12]

total = 0
for s in scores:          # s nimmt nacheinander jeden Wert an
    total = total + s
print("Summe:", total)

for i in range(3):        # 0, 1, 2
    print("Durchlauf", i)

for i, s in enumerate(scores):   # Position und Wert zugleich
    print(i, "->", s)
```

Daneben gibt es `while`, das läuft, solange seine Bedingung stimmt — in diesem Kurs selten, denn meist iterieren wir über Daten, deren Länge feststeht. `break` bricht jede Schleife vorzeitig ab:

```python
power = 1
while power < 100:
    power = power * 2
print(power)             # die erste Zweierpotenz über 100
```

Die Faustregel für die Wahl: direkt über die Elemente, solange Sie die Werte brauchen; `enumerate`, wenn zusätzlich die Position gebraucht wird; eine Index-Schleife nur, wenn Sie wirklich mit Indizes arbeiten müssen, etwa um das rechte Nachbarelement `xs[i + 1]` anzusprechen — dann heißt sie `for i in range(len(xs) - 1)`, denn das letzte Element hat keinen rechten Nachbarn mehr. Auch das ist in Python völlig in Ordnung.

## <a id="kap-funktionen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.8&nbsp;&nbsp;Funktionen</span>

Mit `def` verpacken Sie Logik zum Wiederverwenden. Parameter können Standardwerte haben, `return` gibt das Ergebnis zurück. Und aufgepasst: `print` **zeigt nur an**, `return` **gibt zurück**. Eine Funktion ohne `return` liefert automatisch `None`, egal wie viel sie druckt. Im Beispiel steckt noch eine kleine f-String-Neuheit: Auch die Formatangabe selbst darf Werte einsetzen — bei `digits = 1` wird aus `.{digits}f` die bekannte Angabe `.1f`:

```python
def to_percent(score, digits=1):
    return f"{score * 100:.{digits}f} %"

print(to_percent(0.895))             # nutzt den Standardwert digits=1
print(to_percent(0.895, 0))          # überschreibt ihn über die Position
print(to_percent(0.895, digits=0))   # benanntes Argument: Zuordnung über den Parameternamen
```

Die letzte Variante heißt **benanntes Argument** und macht Aufrufe lesbar, sobald eine Funktion mehrere Parameter hat. Genau so ruft das Workbook `search(3, k=5)` auf.

Mit einem Tupel gibt eine Funktion auch mehrere Werte auf einmal zurück. Links vom Gleichheitszeichen packen Sie das Ergebnis direkt wieder aus:

```python
def score_stats(scores, threshold=0.5):
    # größten Wert und die Anzahl der Werte ab der Schwelle (größer oder gleich) zurückgeben
    best = max(scores)
    hits = 0
    for s in scores:
        if s >= threshold:
            hits += 1
    return best, hits

best, hits = score_stats([0.83, 0.5, 0.77, 0.12])
print(f"bester Wert {best}, {hits} Werte ab der Schwelle")   # die 0.5 zählt mit
```

Nebenbei eine Namensfrage, wie sie in jedem Review auftaucht: Die Funktion hieß im ersten Entwurf `top_score`, aber sie liefert zwei Dinge, nicht eins. Ein Name soll die **ganze** Rückgabe ankündigen, deshalb `score_stats`.

## <a id="kap-variablen-sind-etiketten"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">1.9&nbsp;&nbsp;Variablen sind Etiketten</span>

<div style="border-radius:6px; overflow:hidden; margin:2px 0; margin-right:-20px;">
<div style="background:#C62828; color:#fff; padding:6px 16px; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Definition: Referenzsemantik</div>
<div style="background:#FDECEA; color:#1a1a1a; padding:12px 16px; margin:0;">
<p>Eine Python-Variable ist ein <strong>Name, der auf ein Objekt zeigt</strong>, wie ein Etikett an einem Karton. Eine Zuweisung <code>b = a</code> kopiert nie das Objekt, sondern klebt nur ein zweites Etikett auf denselben Karton.</p>
<details style="margin-top:10px"><summary style="cursor:pointer; color:#C62828; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #C6282855"><p>Zur Einordnung: Bei zwei eigenständigen <code>std::vector&lt;int&gt;</code>-Objekten kopiert die C++-Zuweisung <code>b = a</code> die <strong>Elemente</strong> — die Vektoren lassen sich anschließend unabhängig verändern. Bei Zeigern wird dagegen nur die <strong>Adresse</strong> kopiert, das Zielobjekt bleibt gemeinsam. Die C++-Referenz wiederum bindet <strong>einmal beim Anlegen</strong> und bleibt dann fest. Python bindet bei <strong>jeder</strong> Zuweisung nur den Namen neu an ein Objekt — das kommt der Zeiger-Zuweisung am nächsten, nur ohne eigene Syntax dafür. Das ist der wichtigste Denk-Unterschied zwischen beiden Sprachen und die Quelle der häufigsten Umsteiger-Fehler.</p></div></details>
</div>
</div>

```python
# Zwei Etiketten, ein Karton
a = [1, 2]
b = a              # KEINE Kopie — b zeigt auf dieselbe Liste
b.append(3)
print("a =", a)    # a sieht die Änderung mit
print("b =", b)
print("dasselbe Objekt?", a is b)
```

```python
# Wer wirklich eine Kopie will, sagt es ausdrücklich
a = [1, 2]
b = list(a)        # neue Liste mit denselben Elementen
b.append(3)
print("a =", a, "  b =", b, "  dasselbe Objekt?", a is b)
```

Dasselbe Prinzip entscheidet, was eine Funktion beim Aufrufer anrichtet. Sagen Sie vor dem Ausführen nicht nur voraus, was die beiden `print`-Zeilen zeigen, sondern auch **warum**: Welche der beiden Funktionen verändert den Karton — und welche klebt nur ein neues Etikett auf einen lokalen Namen?

```python
def add_flag(xs):
    xs.append("!")        # verändert das übergebene Objekt — der Aufrufer sieht es

def replace(xs):
    xs = ["neu"]          # bindet nur den LOKALEN Namen neu — der Aufrufer merkt nichts

data = [1, 2]
add_flag(data)
print("nach add_flag:", data)
replace(data)
print("nach replace: ", data)
```

Und noch eine Stufe tiefer: `copy()` (oder `list(a)`) kopiert nur die **äußere** Liste. Stecken darin wieder Listen, zeigen alte und neue Liste auf dieselben inneren Kartons — eine „flache" Kopie. Sagen Sie erst voraus, welche der beiden Listen die `99` am Ende enthält — und begründen Sie, warum:

```python
nested = [[1, 2], [3]]
flat = nested.copy()      # neue äußere Liste …
flat[0].append(99)        # … aber dieselben inneren Listen
print("nested =", nested)
print("flat   =", flat)
```

Der Unterschied zwischen `is` und `==`: `==` fragt „gleicher Inhalt?", `is` fragt „dasselbe Objekt?". Und nicht alles lässt sich überhaupt ändern: Listen sind **veränderlich** (mutable), Zahlen, Strings und Tupel sind **unveränderlich** (immutable): Bei `x = x + 1` entsteht ein neues Objekt, der alte Karton bleibt unberührt. Warum das für NumPy-Arrays und später für GPU-Speicher wichtig wird, sehen Sie im Laufe des Semesters immer wieder.
