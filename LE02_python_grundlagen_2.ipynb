**Tupel oder Liste?** Beide speichern Werte in fester Reihenfolge. Sie werden üblicherweise für verschiedene Zwecke verwendet. Ein Tupel ist ein Datensatz: Jede Position hat eine feste Bedeutung, im Treffer steht vorn der Titel und hinten der Score. Eine Liste enthält gleichartige Werte, deren Anzahl sich ändern kann, zum Beispiel alle Treffer einer Suche. Daraus ergibt sich die Faustregel: Wenn die Position die Bedeutung trägt, passt das Tupel. Wenn Werte hinzukommen oder wegfallen können, passt die Liste. Auch eine Funktion, die mehrere Werte zurückgibt, liefert ein Tupel. Das haben Sie in Teil 1 bei `score_stats` genutzt.

```python
hits = [("Gehirn", 0.89, 92), ("Zeit", 0.72, 157)]     # Titel, Score, Absatznummer
lines = []
for title, score in hits:
    lines.append(f"{score:.2f}  {title}")
print(lines)
```

**Dictionaries.** Ein Dict ordnet Schlüsseln Werte zu. Im Termin haben Sie Werte über den Schlüssel gelesen. Die nächste Zelle zeigt, wie ein Dict verändert und durchlaufen wird.

Bei Listen ist es umgekehrt: Methoden wie `append` verändern die Liste und geben `None` zurück. Das war die `None`-Falle aus Teil 1. Der Unterschied folgt aus den beiden Typen. Ein String lässt sich nicht verändern. Eine String-Methode kann den String deshalb nicht umformen, sie gibt ihr Ergebnis zurück: `lower()` einen neuen String, `split()` eine Liste. Eine Liste lässt sich verändern. Für Methoden, die ihr Objekt nur verändern, gilt in Python die Konvention, `None` zurückzugeben. An der Rückgabe ist so zu erkennen, dass die Methode das vorhandene Objekt verändert hat und kein neues liefert.

```python
def to_word_set(text, clean):
    if clean:
        return set(clean_words(text))            # Satzzeichen an den Wortenden entfernt
    return set(text.lower().split())             # nur zerlegt

def count_alone_on_top(clean):
    passage_sets = []
    for p in passages:
        passage_sets.append(to_word_set(p["text"], clean))
    alone = 0
    for q in queries:
        query_set = to_word_set(q["text"], clean)
        overlaps = []                            # je Absatz die Anzahl gemeinsamer Wörter
        for words in passage_sets:
            overlaps.append(len(query_set & words))
        best = max(overlaps)
        if overlaps[q["expected"]] == best and overlaps.count(best) == 1:
            alone += 1                           # der erwartete Absatz liegt allein vorn
    return alone

print("mit clean_words:       ", count_alone_on_top(True), "von 8")
print("nur lower() und split():", count_alone_on_top(False), "von 8")
q4 = queries[4]
print("Anfrage 4 und ihr Absatz teilen:", shared_words(q4["text"], passages[q4["expected"]]["text"]))
```

Die Stichwortsuche findet den erwarteten Absatz nur bei einem Teil der Anfragen, und das Ergebnis hängt von der Vorverarbeitung ab. Bei der Anfrage 0 („Wo liegt San Diego?") teilt ein Absatz über ein Museum in Mexiko-Stadt ebenso viele Wörter mit der Anfrage wie der gesuchte Absatz, denn dort kommen „Diego Rivera" und „San Angel" vor. Die Anfrage 4 („Auf was beruhen Browserdatenstatistiken?") teilt mit ihrem Absatz kein einziges Wort. Die semantische Suche aus dem Termin hat ihn trotzdem auf Platz 1 gefunden, weil sie Bedeutungen vergleicht und keine Zeichenfolgen.

Die Bereinigung in `clean_words` ist bewusst einfach. Zeichen wie der Apostroph oder der Gedankenstrich bleiben hängen, und „Stadt" und „Städte" gelten als verschiedene Wörter. Für das Zerlegen von Texten gibt es mächtigere Werkzeuge, etwa reguläre Ausdrücke. Sprachmodelle zerlegen Texte mit eigenen Verfahren in Einheiten, die Tokens heißen.

<div style="overflow:hidden; margin-bottom:4px;"><span style="color:#475569; font-weight:bold; font-size:1.08em;">Überblick: Wie übergibt Python Argumente?</span></div>
<p>Python übergibt einer Funktion den Verweis auf das Objekt. Der Parameter ist ein weiterer Name für dasselbe Objekt, das Objekt wird nicht kopiert. Dieses Verfahren heißt <strong>Call by Sharing</strong>.</p>
<p>Falls Sie aus C++ die Begriffe Call by Value und Call by Reference kennen: Python verhält sich wie keine der beiden Übergabearten in C++.</p>
<ul>
<li><strong>Call by Value</strong> heißt in C++: Die Funktion bekommt eine Kopie des Objekts. In Python wird das Objekt nicht kopiert. <code style="color:inherit;">copy_first</code> muss deshalb selbst kopieren.</li>
<li><strong>Call by Reference</strong> heißt in C++: Der Parameter steht für die Variable des Aufrufers, eine Zuweisung an den Parameter ändert diese Variable. In Python lässt eine Zuweisung an den Parameter nur den lokalen Namen auf ein anderes Objekt zeigen.</li>
</ul>
<p>Das <a href="https://docs.python.org/de/3/tutorial/controlflow.html#defining-functions">Python-Tutorial</a> beschreibt dasselbe Verhalten als Call by Value und ergänzt, dass der übergebene Wert immer ein Verweis auf das Objekt ist und nicht das Objekt selbst. Gemeint ist dasselbe wie hier.</p>
<p>Daraus folgen zwei Regeln. Verändert die Funktion das übergebene Objekt, zum Beispiel mit <code style="color:inherit;">records.append(...)</code> oder <code style="color:inherit;">records[0]["title"] = ...</code>, sieht der Aufrufer die Änderung. Weist die Funktion dem Parameter etwas Neues zu, zum Beispiel mit <code style="color:inherit;">records = []</code>, merkt der Aufrufer nichts. Das Beispiel dazu steht in Teil 1 im Kapitel „Variablen sind Etiketten" (<code style="color:inherit;">add_flag</code> und <code style="color:inherit;">replace</code>).</p>
<p>Für Ihre eigenen Funktionen heißt das: Eine Funktion, die ihre Eingabe nicht verändern soll, baut ihr Ergebnis in einer neuen Liste auf und verändert auch die Elemente der Eingabe nicht. Übernimmt sie Elemente ins Ergebnis, gehören diese danach zu beiden Listen. Soll der Aufrufer das Ergebnis frei verändern können, kopiert die Funktion auch die Elemente, wie <code style="color:inherit;">copy_first</code>.</p>

</div>

**`enumerate`.** Aus Teil 1 kennen Sie `enumerate`: Es liefert zu jedem Element seine Nummer. Sind die Elemente selbst Tupel, braucht das Auspacken eine zusätzliche Klammer.

**`zip`.** `zip` durchläuft mehrere Listen gleichzeitig und liefert je Durchlauf ein Tupel aus den Elementen an derselben Position.

```python
titles = ["Gehirn", "Zeit", "Koran"]
scores = [0.89, 0.72, 0.65]
for title, score in zip(titles, scores):
    print(title, score)
print(list(zip(titles, scores)))
print(list(zip(titles, [0.89, 0.72])))
```

**List Comprehensions.** Viele Schleifen bauen aus einer Liste eine neue Liste. Für dieses Muster hat Python eine Kurzform, die List Comprehension. Die nächste Zelle zeigt denselben Rechenweg zweimal.

```python
hits = [("Gehirn", 0.89), ("Zeit", 0.72), ("Koran", 0.65)]

titles = []
for title, score in hits:
    titles.append(title)                 # die Schleife: je Treffer den Titel anhängen
print(titles)

titles = [title for title, score in hits]
print(titles)

good = [title for title, score in hits if score >= 0.7]
print(good)

print([title.upper() for title, score in hits])
```

**Mehrere `for` und `if`.** Eine Comprehension darf mehr als ein `for` und mehr als ein `if` enthalten. Gelesen wird sie von links nach rechts wie verschachtelte Schleifen: Das erste `for` ist die äußere Schleife, das zweite die innere. Ein `if` darf die Namen verwenden, die links davon eingeführt sind, und zwei `if` hintereinander wirken wie eine Bedingung mit `and`. Die nächste Zelle zeigt eine solche Comprehension und darunter dieselbe Rechnung als Schleife.

```python
rows = [[1, 12, 5], [20, 3]]

picked = [x for row in rows for x in row if x > 2 if x < 20]
print(picked)

picked = []
for row in rows:                    # das erste for: die äußere Schleife
    for x in row:                   # das zweite for: die innere Schleife
        if x > 2 and x < 20:        # zwei if hintereinander wirken wie and
            picked.append(x)
print(picked)
```

Eine solche Zeile ist schnell geschrieben und langsam gelesen. Sie ist ein Fall für die Regel oben: Mit verschachtelten Schleifen und mehreren Bedingungen ist die gewöhnliche Schleife leichter zu prüfen.

```python
hits = [("Zeit", 0.72), ("Gehirn", 0.89), ("Koran", 0.65)]

def score_of(hit):
    return hit[1]                      # der Score steht an Position 1 des Paars

print(sorted(hits, key=score_of, reverse=True))

print(sorted(hits, key=lambda hit: hit[1], reverse=True))
```

**Benannte Funktion oder `lambda`?** Beide Zeilen leisten dasselbe. Ein `lambda` eignet sich für einen kurzen Ausdruck, der nur an dieser einen Stelle gebraucht wird. Es kann nur einen einzigen Ausdruck enthalten. Sobald die Funktion mehr tun soll, an mehreren Stellen gebraucht wird oder ein Name das Lesen erleichtert, ist `def` die bessere Wahl.

**`sorted` oder `sort`?** Neben der Funktion `sorted` gibt es die Listen-Methode `sort`. Sie ordnet die Liste selbst um und gibt `None` zurück.

```python
scores = [0.72, 0.89, 0.65]
result = scores.sort()                 # ordnet scores selbst um
print(scores, result)
```

Die Zeile `scores = scores.sort()` ist die `None`-Falle aus Teil 1 in neuer Form: Danach ist `scores` keine Liste mehr. Wichtiger für die Arbeit mit Daten ist der andere Unterschied. `sort` verändert die Liste, auf der es aufgerufen wird. Ist das die Liste, die eine Funktion als Argument bekommen hat, verändert die Funktion die Daten des Aufrufers.

**Gleichstände.** Haben zwei Elemente denselben Schlüssel, behalten sie die Reihenfolge, die sie in der Eingabe hatten. Das gilt auch mit `reverse=True`. Diese Eigenschaft heißt Stabilität.

```python
hits = [("Koran", 0.80), ("Zeit", 0.74), ("Gehirn", 0.80)]
print(sorted(hits, key=lambda hit: hit[1], reverse=True))     # Koran bleibt vor Gehirn
```

Für die Freigabe braucht es deshalb beides. Die Tests decken jede Anforderung ab, auch an den Grenzen. <p>Für mehrere Muster dieses Teils bringt Python fertige Werkzeuge mit. Für diesen Kurs brauchen Sie sie nicht. In fremdem Code begegnen sie Ihnen aber, deshalb zeigt diese Box je ein Beispiel.</p>
<details style="margin-top:10px"><summary style="cursor:pointer; color:#107C72; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #107C7255"><p><strong>Zählen.</strong> <code style="color:inherit;">Counter</code> aus dem Modul <code style="color:inherit;">collections</code> zählt die Elemente einer Liste und liefert die häufigsten:</p>
<pre><code class="language-python">from collections import Counter
counts = Counter([&quot;die&quot;, &quot;gpu&quot;, &quot;die&quot;])
counts[&quot;die&quot;]                  # 2
counts.most_common(1)          # [('die', 2)]
</code></pre>
<p><strong>Gruppieren.</strong> Ein <code style="color:inherit;">defaultdict</code> legt einen fehlenden Eintrag beim ersten Zugriff selbst an. Die Dict-Methode <code style="color:inherit;">setdefault</code> leistet dasselbe für ein gewöhnliches Dict:</p>
<pre><code class="language-python">from collections import defaultdict
by_title = defaultdict(list)                  # fehlender Schlüssel: leere Liste
by_title[&quot;Bern&quot;].append(11)                   # kein KeyError

by_title = {}
by_title.setdefault(&quot;Bern&quot;, []).append(11)    # legt die Liste an, falls der Schlüssel fehlt
</code></pre>
<p><strong>Ein Dict aus zwei Listen.</strong> <code style="color:inherit;">zip</code> liefert die Paare, <code style="color:inherit;">dict</code> macht daraus Einträge:</p>
<pre><code class="language-python">dict(zip([&quot;Gehirn&quot;, &quot;Zeit&quot;], [0.89, 0.72]))   # {'Gehirn': 0.89, 'Zeit': 0.72}
</code></pre>
<p><strong>Auspacken mit Stern.</strong> Ein Name mit Stern nimmt alle übrigen Werte als Liste auf:</p>
<pre><code class="language-python">first, *rest = [1, 2, 3, 4]                   # first ist 1, rest ist [2, 3, 4]
</code></pre>
<p><strong>Zuweisung an einen Slice.</strong> Sie ersetzt den Ausschnitt in der vorhandenen Liste:</p>
<pre><code class="language-python">xs = [0, 1, 2, 3]
xs[1:3] = [9]                                 # xs ist jetzt [0, 9, 3]
</code></pre>
<p><strong><code style="color:inherit;">frozenset</code>.</strong> Ein unveränderliches Set. Es ist hashbar und darf deshalb Schlüssel eines Dicts oder Element eines Sets sein:</p>
<pre><code class="language-python">similarity = {frozenset([&quot;gpu&quot;, &quot;cpu&quot;]): 0.4}
similarity[frozenset([&quot;cpu&quot;, &quot;gpu&quot;])]         # 0.4, die Reihenfolge spielt keine Rolle
</code></pre>
<p><strong>Generatoren.</strong> Eine Funktion mit <code style="color:inherit;">yield</code> liefert ihre Werte einzeln und erst dann, wenn sie abgerufen werden:</p>
<pre><code class="language-python">def squares(n):
    for i in range(n):
        yield i * i

list(squares(4))                              # [0, 1, 4, 9]
</code></pre>
<p><strong><code style="color:inherit;">itertools</code>.</strong> Das Modul enthält Bausteine für Schleifen, zum Beispiel alle Paare aus einer Liste:</p>
<pre><code class="language-python">from itertools import combinations
list(combinations([&quot;a&quot;, &quot;b&quot;, &quot;c&quot;], 2))        # [('a', 'b'), ('a', 'c'), ('b', 'c')]
</code></pre>
<p><strong><code style="color:inherit;">map</code> und <code style="color:inherit;">filter</code>.</strong> Beide leisten dasselbe wie eine Comprehension:</p>
<pre><code class="language-python">words = [&quot;gpu&quot;, &quot;tensor&quot;]
list(map(len, words))                         # [3, 6]
[len(w) for w in words]                       # dasselbe als Comprehension
list(filter(lambda w: len(w) &gt; 3, words))     # ['tensor']
[w for w in words if len(w) &gt; 3]              # dasselbe als Comprehension
</code></pre></div></details>
