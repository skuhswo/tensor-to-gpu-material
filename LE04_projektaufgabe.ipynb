Die Datei `penguins_raw.csv` ist die Rohdatentabelle der Studie von Gorman, Williams und Fraser (Palmer Station, 2007 bis 2009), unverändert aus dem R-Paket palmerpenguins übernommen, Lizenz CC0. Das Datenblatt `DATENBLATT_penguins.md` neben der Datei beschreibt jede Spalte. Die nächste Zelle stellt Daten und Bibliotheken bereit, die Zelle danach zeigt, was in der Datei steckt.

 Laden mit Fehlervertrag -> Vertrag und Bereinigung (N, 8) -> Merkmale und Fragen (Tabellen) -> Analyseblatt (drei Diagramme) -> Array (N, 4): standardisieren, Mittelwerte, nächster Nachbar -> Übergabe: CSV an LE05; über den Kästen die Bibliothek, pandas bis zum Analyseblatt, NumPy für das Array" style="display:block; margin:14px auto; max-width:100%;">

```python
# Kursdaten bereitstellen — nutzt data/, lädt Fehlendes aus dem Kursrepo nach
from pathlib import Path
from urllib.request import urlretrieve
for _name in ["penguins_raw.csv", "DATENBLATT_penguins.md"]:
    _ziel = Path("data") / _name
    if not _ziel.exists():
        _ziel.parent.mkdir(parents=True, exist_ok=True)
        try:
            urlretrieve("https://raw.githubusercontent.com/skuhswo/tensor-to-gpu-material/main/data/" + _name, _ziel)
        except Exception as _fehler:
            _ziel.unlink(missing_ok=True)          # keine halbe Datei zurücklassen
            _tipp = ""
            if "CERTIFICATE" in str(_fehler):      # Python von python.org auf dem Mac
                _tipp = (" Auf dem Mac zuerst 'Install Certificates.command' im Ordner "
                         "Programme/Python ausführen.")
            if _name.endswith(".md"):               # Datenblatt: Lektüre, kein Pflichtteil
                print(_ziel, "nicht nachgeladen (Datenblatt, nur Lektüre):", type(_fehler).__name__)
                continue
            raise RuntimeError(f"{_ziel} fehlt und ließ sich nicht nachladen ({type(_fehler).__name__}). "
                               "Legen Sie den Ordner data/ aus dem ZIP der Lerneinheit neben dieses Notebook."
                               + _tipp) from None
    print(_ziel, "\u2713")
```

Die Vertragsprüfung ist eine Vorgabe. `check_contract(df, contract)` prüft eine Tabelle gegen genau so ein Dict: die Spaltenmenge, die Art der Werte (geprüft an den Werten selbst, nicht am Namen des Datentyps), die zulässigen Werte oder den Bereich und ob Lücken nur dort stehen, wo der Vertrag sie mit `nullable` erlaubt. Lesen Sie sie einmal, sie ist das Gegenstück zur Prüfzelle von `clean_titanic` im Termin, nur mit dem Vertrag als Daten statt als Code.

```python
# Vorgabe: Vertragsprüfung — einfach ausführen
def check_contract(df, contract):
    cols = contract["columns"]
    assert set(df.columns) == set(cols), f"Spalten passen nicht zum Vertrag: {set(df.columns) ^ set(cols)}"
    for col, rule in cols.items():
        s = df[col]
        vals = s.dropna()
        if rule["kind"] == "text":
            assert all(isinstance(v, str) for v in vals), f"{col}: Text erwartet"
        elif rule["kind"] == "bool":
            assert all(isinstance(v, (bool, np.bool_)) for v in vals), f"{col}: True/False erwartet"
        elif rule["kind"] == "int":
            assert s.dtype.kind == "i", f"{col}: ganzzahliger Datentyp erwartet, ist {s.dtype}"
        elif rule["kind"] == "float":
            assert s.dtype.kind == "f", f"{col}: Fließkomma-Datentyp erwartet, ist {s.dtype}"
        else:
            raise ValueError(f"unknown kind {rule['kind']!r} for column {col!r}")
        if "values" in rule:
            assert set(vals) <= set(rule["values"]), f"{col}: unzulässige Werte {set(vals) - set(rule['values'])}"
        if "range" in rule:
            lo, hi = rule["range"]
            assert vals.between(lo, hi).all(), f"{col}: Werte außerhalb von {lo} bis {hi}"
        if not rule.get("nullable", False):
            assert s.notna().all(), f"{col}: Lücken sind laut Vertrag nicht erlaubt, es gibt {s.isna().sum()}"
    return True

print("check_contract bereit")
```
