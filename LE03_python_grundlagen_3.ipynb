## <a id="kap-dateien-und-json"></a><span class="zwischentitel" style="display:block; color:#0E2841; color:var(--vscode-foreground, #0E2841); border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;">3.1&nbsp;&nbsp;Dateien und JSON</span>

Zuerst die Daten. Es sind dieselben wie im Termin zu LE03: die 400 Absätze und die acht Beispielanfragen als JSON, dazu die zugehörigen Vektoren als `.npy`-Dateien. Die nächste Zelle legt sie in den Ordner `data/`, falls sie dort fehlen.

```python
# Kursdaten bereitstellen — nutzt data/, lädt Fehlendes aus dem Kursrepo nach
from pathlib import Path
from urllib.request import urlretrieve
for _name in ["appetizer_passages.json", "appetizer_queries.json", "appetizer_embeddings.npy", "appetizer_query_embeddings.npy"]:
    _ziel = Path("data") / _name
    if not _ziel.exists():
        _ziel.parent.mkdir(parents=True, exist_ok=True)
        try:
            urlretrieve("https://raw.githubusercontent.com/skuhswo/tensor-to-gpu-material/main/data/" + _name, _ziel)
        except Exception as _fehler:
            _ziel.unlink(missing_ok=True)          # keine halbe Datei zurücklassen
            _tipp = ""
            if "CERTIFICATE" in str(_fehler):      # Python von python.org auf dem Mac
                _tipp = (" Auf dem Mac zuerst 'Install Certificates.command' im Ordner "
                         "Programme/Python ausführen.")
            raise RuntimeError(f"{_ziel} fehlt und ließ sich nicht nachladen ({type(_fehler).__name__}). "
                               "Legen Sie den Ordner data/ aus dem ZIP der Lerneinheit neben dieses Notebook."
                               + _tipp) from None
    print(_ziel, "\u2713")
```

<details style="margin-top:10px"><summary style="cursor:pointer; color:#E97132; font-weight:600; font-size:0.9em">Antwort anzeigen</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #E9713255"><p>Einen leeren String <code style="color:inherit;">""</code>. <div style="background:#2E75B6; color:#fff; padding:6px 16px; box-sizing:border-box; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Genau verstehen: Warum with?</div>
<div style="background:#EAF2FA; color:#1a1a1a; padding:12px 16px; box-sizing:border-box; margin:0;">
<p><code style="color:inherit;">open</code> gibt ein Dateiobjekt zurück, und eine geöffnete Datei belegt Betriebsmittel, bis sie geschlossen wird: einen Dateizugriff beim Betriebssystem und einen Puffer im Speicher. Ohne <code style="color:inherit;">with</code> müssten Sie <code style="color:inherit;">f.close()</code> selbst aufrufen, und zwar auf jedem Ausführungspfad aus dem Code heraus, auch dann, wenn zwischen <code style="color:inherit;">open</code> und <code style="color:inherit;">close</code> ein Fehler auftritt. Der <code style="color:inherit;">with</code>-Block übernimmt das automatisch: Am Ende des eingerückten Blocks wird die Datei geschlossen, auch wenn der Block mit einer Ausnahme verlassen wird. Beim Schreiben ist das wichtig, weil <code style="color:inherit;">write</code> die Daten zunächst in einem Puffer sammelt und erst das Schließen sie vollständig in die Datei bringt. Eine Datei, die nicht geschlossen wurde, kann unvollständig auf der Platte liegen.</p>
<p>Das Muster ist deshalb immer dasselbe: <code style="color:inherit;">with open(...) as f:</code> und darunter die Arbeit mit <code style="color:inherit;">f</code>. Nach dem Block ist <code style="color:inherit;">f</code> geschlossen, ein weiteres <code style="color:inherit;">f.read()</code> bricht mit einem <code style="color:inherit;">ValueError: I/O operation on closed file</code> ab. Die Mechanik dahinter heißt Context Manager: ein Objekt, das beim Betreten des <code style="color:inherit;">with</code>-Blocks eine Vorbereitung ausführt und beim Verlassen ein Aufräumen, bei einer Datei das Schließen. Eigene Context Manager gehören zum Material Advanced Python.</p>
<p>Was <code style="color:inherit;">with</code> nicht tut: Schreibvorgänge rückgängig machen. Scheitert <code style="color:inherit;">json.dump</code> mitten in der Datei, etwa an einem NumPy-Array, ist die Datei danach zwar geschlossen, aber der alte Inhalt ist durch den Modus <code style="color:inherit;">"w"</code> schon gelöscht, und ein unvollständiger Anfang kann stehen bleiben. Auch <code style="color:inherit;">try</code> und <code style="color:inherit;">except</code> setzen nichts zurück.</p>
<p><strong>Die Zeichenkodierung.</strong> <code style="color:inherit;">encoding="utf-8"</code> steht in jeder <code style="color:inherit;">open</code>-Zeile dieses Kurses. Eine Datei speichert Bytes, und die Kodierung legt fest, welche Bytes welches Zeichen bedeuten. Ohne Angabe hängt die Kodierung von der Python-Version und von der Umgebung ab: vor Python 3.15 häufig von den Einstellungen des Betriebssystems, ab 3.15 ist UTF-8 die Voreinstellung. Wir geben sie an, weil die Kursdateien in UTF-8 gespeichert sind und der Code unabhängig von der Voreinstellung funktionieren soll. Mit einer falschen Kodierung erscheinen Umlaute als unlesbare Zeichenfolgen, oder das Lesen bricht mit einem <code style="color:inherit;">UnicodeDecodeError</code> ab.</p>

**JSON.** Die Kursdaten liegen als JSON vor, einem Textformat für strukturierte Daten.

<div style="border-radius:6px; overflow:hidden; margin:2px 0; margin-right:-20px; margin-right:calc(var(--jp-cell-padding, 0px) * -4);">
<div style="background:#2E75B6; color:#fff; padding:6px 16px; box-sizing:border-box; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Genau verstehen: Das Format JSON</div>
<div style="background:#EAF2FA; color:#1a1a1a; padding:12px 16px; box-sizing:border-box; margin:0;">
<p>JSON (JavaScript Object Notation) ist Text, der Daten nach sechs Regeln aufschreibt: Ein <strong>Objekt</strong> ist eine Zuordnung in geschweiften Klammern, <code style="color:inherit;">{"id": 6, "title": "Gehirn"}</code>, die Schlüssel sind immer Strings. Ein <strong>Array</strong> ist eine Folge in eckigen Klammern, <code style="color:inherit;">[1, 2, 3]</code>. Ein <strong>String</strong> steht in doppelten Anführungszeichen, eine <strong>Zahl</strong> steht nackt (<code style="color:inherit;">92</code>, <code style="color:inherit;">0.89</code>), dazu kommen <code style="color:inherit;">true</code>, <code style="color:inherit;">false</code> und <code style="color:inherit;">null</code>. Objekte und Arrays dürfen beliebig ineinander geschachtelt sein, Leerzeichen und Zeilenumbrüche zwischen den Teilen spielen keine Rolle. Mehr gibt es nicht; genau deshalb können fast alle Sprachen und Werkzeuge JSON lesen.</p>
<p>Die Datei <code style="color:inherit;">appetizer_passages.json</code> ist nach diesem Muster gebaut, hier schematisch, auf den Schlüssel <code style="color:inherit;">passages</code> reduziert und mit erfundenen Beispieldaten:</p>
<pre><code style="color:inherit;">{&quot;passages&quot;: [{&quot;id&quot;: 6, &quot;title&quot;: &quot;Gehirn&quot;, &quot;text&quot;: &quot;Das Gehirn ...&quot;},
              {&quot;id&quot;: 7, &quot;title&quot;: &quot;Gehirn&quot;, &quot;text&quot;: &quot;...&quot;}, ...]}
</code></pre>
<p>Die Entsprechungen sind naheliegend: Objekt ist Dict, Array ist Liste, der Rest sind String, Zahl, Wahrheitswert und <code style="color:inherit;">None</code>.</p>

</div>
</div>

Das Modul `json` übersetzt zwischen diesem Text und Python-Objekten: `json.load` liest aus einer geöffneten Datei, `json.dump` schreibt in eine. Die nächste Zelle schreibt ein kleines Dict, zeigt den Dateiinhalt und liest ihn zurück.

```python
import json

record = {"title": "Gehirn", "hit": ("Gehirn", 0.89), "ids": [92, 157], 7: "sieben", "checked": True, "note": None}
with open("out/record.json", "w", encoding="utf-8") as f:
    json.dump(record, f, indent=2, ensure_ascii=False)

print("Dateiinhalt:")
with open("out/record.json", encoding="utf-8") as f:
    print(f.read())

with open("out/record.json", encoding="utf-8") as f:
    restored = json.load(f)
print("Zurückgelesen:", restored)
print("Gleich dem Original?", restored == record)
```

Die Rückübersetzung ist nicht in jedem Fall identisch mit dem Original, deshalb ist der Vergleich am Ende `False`. JSON kennt kein Tupel, aus `("Gehirn", 0.89)` wird eine Liste. Und die Schlüssel eines JSON-Objekts sind immer Strings, aus dem Schlüssel `7` wird `"7"`. Für die Kursdaten spielt das keine Rolle, denn dort sind alle Schlüssel Strings und alle Folgen Listen. Für eigene Daten heißt es: Was aus JSON zurückkommt, sind Dicts mit String-Schlüsseln, Listen, Strings, Zahlen, Wahrheitswerte und `None`. Objekte anderer Typen, etwa ein NumPy-Array, lassen sich nicht direkt schreiben; `json.dump` bricht dann mit einem `TypeError` ab.

Eine JSON-Datei enthält genau ein Dokument. Zwei `json.dump`-Aufrufe hintereinander ergeben etwa `{"a": 1}{"b": 2}`, und `json.load` bricht damit ab (`JSONDecodeError: Extra data`). Mehrere Datensätze kommen als eine Liste in einen Aufruf. Das Format JSON Lines, eine Zeile je Datensatz, ist ein eigenes Format mit eigenen Werkzeugen.

- die Datei fehlt,

Führen Sie beide Zellen einfach aus.

```python
# Vorgabe: load_passages wie in der Musterlösung von Aufgabe 3.1 #1 — einfach ausführen
import json
from pathlib import Path

def load_passages(path):
    """Liest die Liste der Absätze aus einer JSON-Datei im Format der Kursdaten."""
    with open(path, encoding="utf-8") as f:
        return json.load(f)["passages"]

Path("out").mkdir(exist_ok=True)
```

**Abfangen mit `try` und `except`.** Bisher hat jede Ausnahme die Zelle beendet: Python hat die Ausführung abgebrochen und den Traceback ausgegeben. Mit `try` und `except` behandeln Sie eine Ausnahme stattdessen selbst. Der `try`-Block enthält den Code, der scheitern kann; der `except`-Block legt fest, was in diesem Fall geschehen soll, etwa eine Meldung ausgeben und mit einem Ersatzwert weitermachen. Die nächste Zelle lädt die Daten über eine Funktion, die die beiden Fehlerfälle meldet und dann mit einer leeren Liste weitermacht.

```python
def try_load(path):
    try:
        passages = load_passages(path)
    except FileNotFoundError as e:
        print("Datei fehlt:", e)
        return []
    except ValueError as e:
        print("Inhalt ungültig:", e)
        return []
    print(len(passages), "Absätze aus", path)
    return passages

try_load("data/appetizer_passages.json")
try_load("data/missing.json")
result = try_load("out/broken.json")
print(result)                                   # die leere Liste nach dem abgefangenen Fehler
```

Der Ablauf ist immer derselbe. Python führt den `try`-Block aus. Entsteht darin eine Ausnahme, wird der Rest des Blocks übersprungen, und Python sucht den ersten `except`-Block, dessen Klasse zur Ausnahme passt. Danach geht es nach dem ganzen `try`-Konstrukt weiter. Passt kein `except`, wandert die Ausnahme zum Aufrufer, und so weiter, bis sie jemand abfängt oder die Zelle endet. Dadurch entsteht ein Traceback wie im Beispiel oben. Entsteht keine Ausnahme, werden die `except`-Blöcke übersprungen.

Zwei Schreibweisen ergänzen das Muster: `except (FileNotFoundError, ValueError):` fängt mehrere Klassen mit demselben Block, und `except ValueError:` ohne `as e` genügt, wenn die Meldung nicht gebraucht wird. In C++ entspricht `except` dem `catch`, und `raise` weiter unten dem `throw`.

**`else` und `finally`.** Zwei weitere Blöcke können an ein `try` anschließen. Ein `else`-Block läuft nur, wenn im `try`-Block keine Ausnahme aufgetreten ist; dorthin gehört der Code, der das Ergebnis weiterverwendet, damit nicht auch seine Fehler vom `except` abgefangen werden. Ein `finally`-Block läuft immer, mit und ohne Ausnahme, auch wenn ein `except` die Funktion per `return` verlässt; er ist für Aufräumarbeiten gedacht, etwa eine Verbindung zu schließen. Für Dateien übernimmt das schon der `with`-Block.

```python
def try_load_report(path):
    try:
        passages = load_passages(path)
    except FileNotFoundError as e:
        print("Datei fehlt:", e)
    else:
        print(len(passages), "Absätze aus", path)
    finally:
        print("Versuch beendet:", path)

try_load_report("data/appetizer_passages.json")
try_load_report("data/missing.json")
```

```python
def score_label(score):
    if not 0 <= score <= 1:
        raise ValueError(f"score out of range: {score}")
    if score >= 0.7:
        return "gut"
    return "schwach"

print(score_label(0.89))
try:
    print(score_label(1.5))
except ValueError as e:
    print("Abgefangen:", e)
```

Die Fehlerklasse wählen Sie nach diesem Überblick: `ValueError` für einen unzulässigen Wert, `TypeError` für einen falschen Typ, `KeyError` oder `FileNotFoundError`, wenn Ihre Funktion dieselbe Situation meldet wie die eingebauten. Die Meldung nennt den erwarteten und den tatsächlichen Wert, denn der Aufrufer kennt den Code Ihrer Funktion nicht und muss allein aus der Meldung verstehen, was falsch war. Die Meldung schreiben wir wie den übrigen Code auf Englisch: Sie steht im Traceback neben Pythons eigenen Meldungen, und die sind englisch.

**`assert` oder `raise`?** Beide brechen ab, sie haben aber verschiedene Aufgaben. `assert` prüft eine Annahme, die im Programm gelten muss, und ist für Tests gedacht, so wie in den Prüfzellen dieses Kurses. Python kann alle `assert`-Anweisungen abschalten, wenn es mit der Option `-O` läuft, deshalb darf eine Funktion ihre Eingaben nicht mit `assert` prüfen. Eingaben prüft die Funktion selbst mit `if` und löst bei einer verletzten Bedingung mit `raise` eine Ausnahme aus. `check_input` ist ein Beispiel hierfür.

**Eigene Fehlerklassen.** Wenn die eingebauten Klassen nicht treffen oder ein Programm seine eigenen Fehlerarten unterscheiden will, definiert es eine Unterklasse von `Exception`. Der Rumpf braucht nur einen Docstring, geerbt wird alles. Aufrufer können diese Klasse gezielt abfangen, ohne andere Fehler mitzufangen. Wer sie von `ValueError` statt von `Exception` ableitet, erreicht, dass auch ein allgemeines `except ValueError` sie noch fängt. Im Grundkurs reichen die eingebauten Klassen; der Fehlervertrag unten verwendet `ValueError`.

```python
class CorpusError(Exception):
    """Die Kursdaten haben nicht das erwartete Format."""

def passages_of(data, path):
    if "passages" not in data:
        raise CorpusError(f"key 'passages' missing in {path}")
    return data["passages"]

try:
    passages_of({"items": []}, "out/no_key.json")
except CorpusError as e:
    print("Kursdaten unbrauchbar:", e)
```

**Ein Fehlervertrag für die Ladefunktion.** Der Aufrufer von `load_passages` muss wissen, welche Fehler sie melden kann, sonst lässt sich kein passendes `except` schreiben. Für die drei Fälle legen wir fest:

2. **Der Inhalt ist kein gültiges JSON:** `ValueError`, so wie `json.load` ihn auslöst (als `JSONDecodeError`). Auch er wird unverändert weitergereicht.
3. **Der Schlüssel `passages` fehlt:** `ValueError` mit einer eigenen Meldung, die den Schlüssel und den Pfad nennt. Ohne diese Prüfung meldet die Funktion einen `KeyError: 'passages'`, der nicht sagt, welche Datei betroffen ist.

Für die ersten beiden Fälle ist nichts zu tun, die eingebauten Fehler sind schon verständlich. Der dritte Fall ist ein Fall für Abfragen: `"passages" in data` ist billig und vollständig.

Der Vertrag prüft die JSON-Syntax und den Schlüssel, mehr nicht. `{"passages": 42}` ist gültiges JSON mit dem Schlüssel, und die Funktion gibt dann `42` zurück. Dass der Wert eine Liste von Absätzen mit den erwarteten Schlüsseln ist, setzen wir voraus, so wie die Prüfzellen in LE03 die Kursdaten voraussetzen. Die Zusatzaufgabe zieht die Grenze einen Schritt weiter.

Eine gute Stelle, den Vertrag festzuhalten, ist der Docstring der Funktion, mit einem Satz je Fehlerfall; wer `help(load_passages)` aufruft, sieht ihn dann. Die Musterlösung und die Fassung in `tools.py` im nächsten Kapitel machen es so.

Welche Form passt? `import m` hält die Herkunft sichtbar: Bei `np.sqrt` ist zu sehen, dass NumPy rechnet. `from m import f` passt für einzelne, häufig gebrauchte Namen. `from m import *` holt alle Namen des Moduls ungesehen ins Notebook; dann ist nicht mehr erkennbar, woher ein Name stammt, und gleichnamige Funktionen überschreiben sich. Diese Form kommt im Kurs nicht vor und wird allgemein nicht empfohlen.

**Wo Python sucht.** Beim `import` geht Python die Liste `sys.path` der Reihe nach durch, eine gewöhnliche Python-Liste von Ordnerpfaden im Modul `sys` (`import sys` und `print(sys.path)` zeigen sie). Im Notebook steht darin das Arbeitsverzeichnis des Kernels, dazu die installierte Umgebung, also der Ort, an dem `uv` oder Colab die Pakete abgelegt haben (siehe das Begleitdokument zur Entwicklungsumgebung). Eine eigene Datei `tools.py` im Arbeitsverzeichnis ist damit importierbar, und `tools.__file__` zeigt, welche Datei geladen wurde. Daraus folgt eine Regel: Eigene Dateien nie `json.py` oder `numpy.py` nennen, sie verdecken sonst die Bibliothek. Findet Python kein Modul mit dem Namen, meldet es einen `ModuleNotFoundError`; gibt es das Modul, aber nicht den Namen in `from m import name`, einen `ImportError`.

**Ein eigenes Modul anlegen.** Ein eigenes Modul ist nichts anderes als eine `.py`-Datei neben dem Notebook; Sie könnten `tools.py` mit einem Editor anlegen und darin Funktionen definieren. Im Notebook geht es ohne Editor: Der Zellbefehl `%%writefile tools.py` schreibt den Inhalt der Zelle in diese Datei, statt ihn als Python auszuführen. Die Datei landet im Arbeitsverzeichnis des Kernels, und der Befehl überschreibt eine vorhandene Datei vollständig. Die nächste Zelle legt so das Modul mit der Funktion `clean_words` aus Teil 2 an, die Zelle danach importiert es.

```python
%%writefile tools.py
"""Hilfsfunktionen für die Suche: Text bereinigen, Kursdaten laden."""
import json

PUNCTUATION = ".,;:!?()\"„“"


def clean_words(text):
    """Gibt die Wörter des Textes in Kleinbuchstaben und ohne Satzzeichen an den Wortenden zurück."""
    result = []
    for word in text.lower().split():
        word = word.strip(PUNCTUATION)
        if word != "":
            result.append(word)
    return result
```

Zwei Dinge sind dabei wichtig. Erstens hat das Modul seinen eigenen Namensraum: `tools.py` muss `json` selbst importieren, wenn eine seiner Funktionen `json.load` aufruft. Ein `import json` im Notebook gilt nur im Notebook. Deshalb steht die Zeile `import json` oben in der Datei, obwohl `clean_words` sie aktuell noch nicht braucht; die nächste Aufgabe braucht sie dann. Zweitens führt Python die Datei beim ersten `import` aus, von oben nach unten: Jedes `def` legt eine Funktion an, ein `print` auf Modulebene gibt etwas aus. Das Ergebnis bleibt im Speicher, ein erneutes `import tools` führt nichts mehr aus. `import` ist also keine Textersetzung wie `#include` in C. Ändern Sie `tools.py`, laden Sie das Modul mit `importlib.reload(tools)` neu, oder Sie starten den Kernel neu. Die nächsten beiden Zellen zeigen das an einer Minidatei.

Beim Import heißt das Modul wie die Datei, `__name__` ist `"demo"`, und der Block unter `if __name__ == "__main__":` wird übersprungen. Starten Sie die Datei dagegen direkt, mit `python demo.py` im Terminal, ist `__name__` gleich `"__main__"`, und der geschützte Block läuft mit. So trennt die Zeile am Ende vieler Python-Dateien einen Testaufruf, der nur beim direkten Start laufen soll, von den Definitionen, die der Import bereitstellt. Ein Paket ist ein Ordner mit Modulen, also mit `.py`-Dateien; `numpy` ist eines, und `np.linalg` ist ein Modul darin.

Noch ein Hinweis zu `reload`: Es erneuert das Modul, nicht die Namen, die Sie vorher mit `from tools import …` ins Notebook geholt haben; die zeigen weiter auf die alte Fassung. Solange Sie an einem Modul arbeiten, greifen Sie deshalb über `tools.` zu oder wiederholen den `from`-Import nach jedem `reload`.

Die Ausnahme aus `float` wird in `read_scores` abgefangen und durch eine eigene ersetzt, die mehr weiß als `float`: die Zeilennummer. Der Aufrufer bekommt die neue Ausnahme; im Traceback erscheint die ursprüngliche darüber mit dem Vermerk „During handling of the above exception, another exception occurred". Das ist beabsichtigt, beide Meldungen zusammen erklären den Fehler.

Zum Lesen fremden Codes drei Formen des Weiterreichens: Ohne `except` wandert eine Ausnahme von selbst zum Aufrufer. Ein bloßes `raise` in einem `except`-Block löst die gerade behandelte Ausnahme erneut aus, etwa nach einer Protokollausgabe. `raise ValueError(...) from e` ersetzt sie wie in `read_scores`, nennt die Ursache aber ausdrücklich; im Traceback steht dann „The above exception was the direct cause of the following exception".
