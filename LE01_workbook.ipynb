Eine Suchmaschine, die Bedeutung versteht, und der Blick auf die sieben Ebenen, auf denen sie rechnet. Sie geben eine Frage ein und bekommen die passenden Wikipedia-Absätze, obwohl kein einziges Wort der Frage darin vorkommen muss. Wie das funktioniert, ist ein durchgängiges Beispiel in diesem Modul. Heute schauen Sie nur zu, drehen an den Eingaben und sehen, wie unterschiedlich schnell dieselbe Rechnung auf sieben Ebenen läuft. Die Python-Konstrukte in den Zellen müssen Sie noch nicht selbst schreiben können — die Kreisziffern erklären sie, und geübt wird im Lab.

Unsere Suchmaschine besteht aus zwei Zutaten, die im Ordner `data/` liegen und die die nächste Zelle bei Bedarf automatisch nachlädt:

- **400 Absätze aus der deutschen Wikipedia**: unser Dokumentbestand, der „Korpus". Jeder Absatz stammt aus einem anderen Artikel.
- **Je Absatz ein Vektor aus 384 Zahlen**: die sogenannten Embeddings. Ein Sprachmodell hat jeden Absatz vorab in einen solchen Zahlenvektor übersetzt; die Idee dahinter zeigt der Abschnitt <a href="#kap-wie-kommt-bedeutung-in-zahlen">1.1.2</a>.

Dazu kommen acht **Beispielanfragen** (passende Fragen zu diesem Korpus), ebenfalls schon als Text und als Vektor. So funktioniert die Suche auch ohne installiertes Sprachmodell.

````python
# Kursdaten bereitstellen — nutzt data/, lädt Fehlendes aus dem Kursrepo nach
from pathlib import Path
from urllib.request import urlretrieve
for _name in ["appetizer_passages.json", "appetizer_embeddings.npy", "appetizer_queries.json", "appetizer_query_embeddings.npy", "appetizer_timings.json"]:
    _ziel = Path("data") / _name
    if not _ziel.exists():
        _ziel.parent.mkdir(parents=True, exist_ok=True)
        urlretrieve("https://raw.githubusercontent.com/skuhswo/tensor-to-gpu-material/main/data/" + _name, _ziel)
    print(_ziel, "\u2713")
````

````python
import json
import numpy as np

# Korpus laden: die Absätze als Text und die zugehörigen Vektoren als Matrix.
# E hat die Form [400, 384]: eine Zeile je Absatz, 384 Zahlen je Zeile.
corpus = json.load(open("data/appetizer_passages.json"))
passages = corpus["passages"]
E = np.load("data/appetizer_embeddings.npy")

# Die Beispielanfragen: ebenfalls Text plus je ein Vektor (Matrix Q, Form [8, 384]).
queries = json.load(open("data/appetizer_queries.json"))["queries"]
Q = np.load("data/appetizer_query_embeddings.npy")

print(len(passages), "Passagen ·", E.shape[1], "Zahlen je Vektor ·", len(queries), "Beispielanfragen")
````

Texte mit ähnlicher Bedeutung bekommen vom Sprachmodell ähnliche Vektoren: „Wo liegt San Diego?" und ein Absatz über die Geographie San Diegos zeigen im 384-dimensionalen Raum in fast dieselbe Richtung. Suchen heißt deshalb nur noch, den Vektor der Frage mit allen 400 Absatz-Vektoren zu vergleichen und die ähnlichsten herauszusuchen. Drei Schritte: **Normalize** (alle Vektoren auf Länge 1 bringen, damit nur die Richtung zählt, nicht die Länge), **Compare** (Skalarprodukt des Frage-Vektors mit jedem Absatz-Vektor; bei Länge 1 ist das genau die Kosinus-Ähnlichkeit: 1 = gleiche Richtung, 0 = senkrecht zueinander), **Select** (die k besten Treffer nehmen).

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Warum das Normalisieren dazugehört: Ohne diesen Schritt würde die Länge der Vektoren das Ergebnis verzerren, und ein Absatz könnte allein durch betragsmäßig große Zahlen „ähnlicher" wirken als sein Inhalt rechtfertigt. Nach dem Normalisieren ist das Skalarprodukt genau der Kosinus des Winkels zwischen den beiden Richtungen, und nur die Richtung kodiert die Bedeutung. Das unterscheidet diese Suche auch von der klassischen Stichwortsuche: Verglichen werden nicht Buchstabenfolgen, sondern Bedeutungen. „Wo liegt San Diego?" findet deshalb auch Absätze, die keines der Frageworte enthalten, solange sie inhaltlich passen. Genau diese drei Schritte begleiten Sie als Semesterfall durch alle Lerneinheiten: erst in NumPy, später in PyTorch, am Ende als eigener GPU-Kernel. Ein erstes intuitives Bild, wie ein Sprachmodell zu seinen Vektoren kommt und was die Kosinus-Ähnlichkeit misst, gibt gleich der Abschnitt <a href="#kap-wie-kommt-bedeutung-in-zahlen">1.1.2</a>.</p></div></details>

Die folgende Zelle setzt diese drei Schritte in wenigen Zeilen um. Die Funktion `search(query_no, k)` hat zwei Parameter: `query_no` ist die **Nummer der Beispielanfrage** aus der Liste oben (0 bis 7), und `k` bestimmt, **wie viele Treffer** angezeigt werden. Hinter jedem Treffer steht sein Ähnlichkeitswert. Je näher an 1, desto besser passt der Absatz zur Frage:

````python
def normalize(v):
    # Jede Zeile durch ihre Länge teilen -> alle Vektoren haben Länge 1
    return v / np.linalg.norm(v, axis=-1, keepdims=True)

def search(query_no, k=3):
    q = normalize(Q[query_no])               # Normalize: der Fragevektor
    scores = normalize(E) @ q                # Compare: Kosinus-Ähnlichkeit zu allen 400 Absätzen
    best = np.argsort(scores)[::-1][:k]      # Select: die k höchsten Werte
    print("Anfrage:", queries[query_no]["text"], "\n")
    for pid in best:
        p = passages[pid]
        print(f"  {scores[pid]:.3f}  {p['title']}")
        print("         " + p["text"][:160].replace("\n", " ") + " …\n")

# Anfrage Nr. 0 ("Wo liegt San Diego?"), drei Treffer:
search(0)
````

### <a id="kap-wie-kommt-bedeutung-in-zahlen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.1.2&nbsp;&nbsp;Wie kommt Bedeutung in Zahlen?</span>

Die Suche eben hat mit „Vektoren aus 384 Zahlen" gerechnet. Zeit für die Frage, die sich dabei aufdrängt: Wie kommt die **Bedeutung** eines Textes in Zahlen? Ein neuronales Netz rechnet mit Zahlen, nicht mit Buchstaben. Mit dem Satz „San Diego liegt am Pazifik" kann es zunächst nichts anfangen. Irgendeine Übersetzung von Text in Zahlen muss also her, und wie man sie wählt, entscheidet darüber, was das Netz lernen kann. Ein intuitives Bild davon genügt für heute.

Der naheliegende erste Versuch: ein Vokabular nehmen (sagen wir den Duden) und die Wörter durchnummerieren. „Hund" bekommt die 2.317, „Katze" die 3.589. Damit das Netz die Nummern nicht fälschlich als Größen versteht (3.589 ist nicht „mehr" als 2.317), schreibt man sie meist als **One-Hot-Vektor**: ein Vektor, so lang wie das Vokabular, mit einer 1 an der Position des Wortes und sonst lauter Nullen. Das funktioniert, hat aber zwei Probleme. Die Vektoren sind riesig (eine Position je Wort im Vokabular). Und vor allem: Sie enthalten keinerlei Bedeutung. „Hund" und „Katze" sind in dieser Darstellung genauso verschieden wie „Hund" und „Auto" — jedes Wort steht allein. Gesucht ist also eine Darstellung, in der ähnliche Bedeutung auch ähnliche Zahlen ergibt.

Woher soll eine Maschine wissen, was ein Wort bedeutet? Die Antwort, auf der moderne Sprachmodelle beruhen, ist verblüffend einfach: **aus dem Kontext**. Schauen Sie, was in diese Lücken passt:

- „Ich trinke morgens eine Tasse ___." — *Tee*, *Kaffee* (aber nicht *Auto*)
- „Der ___ bellt jede Nacht im Garten." — *Hund* (vielleicht noch *Wachhund*, sicher keine *Katze*)
- „Wir fahren mit dem ___ in den Urlaub." — *Auto*, *Zug*, *Wohnmobil*

Wörter, die in denselben Lücken stehen können, also in ähnlichen Kontexten auftauchen, haben ähnliche Bedeutung. Ein Sprachmodell liest Milliarden von Sätzen und lernt daraus für jedes Wort (und zusammengesetzt für ganze Absätze) einen Zahlenvektor, sodass ähnliche Kontexte zu ähnlichen Vektoren führen. Bedeutung steckt dann nicht in einer einzelnen Zahl, sondern in der **Richtung**, in die der Vektor zeigt.

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Die Idee ist älter als die heutige KI: „You shall know a word by the company it keeps" (J. R. Firth, 1957). Eine einzelne Zahl je Wort würde übrigens nicht reichen, denn Bedeutung hat viele Facetten (Lebewesen oder Gegenstand? konkret oder abstrakt? positiv oder negativ?). In echten Embeddings liegt eine solche Facette allerdings nicht in einer einzelnen, benennbaren Dimension, sondern verteilt über viele Koordinaten zugleich — auf „die Obst-Dimension" kann im Vektor niemand zeigen. Als stark vereinfachtes Denkbild hilft sie trotzdem: Ein mehrdeutiges Wort wie „Apple" braucht Anteile in Richtung Obst und in Richtung Firma. Moderne Sprachmodelle lösen Mehrdeutigkeit zusätzlich über den Kontext auf: Sie berechnen den Vektor nicht je Wort, sondern je Verwendung — „Apple" im Obstsalat-Satz bekommt einen anderen Vektor als „Apple" in der Quartalsbilanz.</p></div></details>

<div style="border-radius:6px; overflow:hidden; margin:2px 0; margin-right:-20px;">
<div style="background:#C62828; color:#fff; padding:6px 16px; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Definition: Embedding</div>
<div style="background:#FDECEA; color:#1a1a1a; padding:12px 16px; margin:0;">
<p>Ein <strong>Embedding</strong> ist ein gelernter Zahlenvektor, der ein Objekt — ein Wort, einen Satz, einen Absatz — so in einem Raum platziert, dass <strong>ähnliche Bedeutung zu ähnlichen Vektoren</strong> führt. Typisch sind einige hundert Dimensionen; unser Kursmodell verwendet 384.</p>
<details style="margin-top:10px"><summary style="cursor:pointer; color:#C62828; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #C6282855"><p>In diesem Raum lässt sich sogar rechnen: Das berühmteste Beispiel der Embedding-Forschung ist <em>König − Mann + Frau ≈ Königin</em>. Zieht man vom König-Vektor den Mann-Vektor ab, bleibt so etwas wie „Herrscher ohne Geschlecht" übrig; addiert man den Frau-Vektor, landet man in der Nähe von „Königin". Die Richtung „männlich → weiblich" ist im Vektorraum also tatsächlich eine Richtung, und sie ist für König/Königin dieselbe wie für Mann/Frau. Wie solche Strukturen beim Lernen entstehen, zeigt anschaulich die Vertiefung am Ende dieses Abschnitts.</p></div></details>
</div>
</div>

Ein Spielzeugbeispiel mit nur vier Dimensionen macht das greifbar. Wir geben drei Wörtern von Hand kleine Vektoren (*Hund* und *Katze* bekommen ähnliche Zahlen, *Auto* ganz andere) und messen die Ähnlichkeit dann genauso, wie es die Suche oben tut: mit der **Kosinus-Ähnlichkeit**.

````python
# Spielzeug-Embeddings: vier Dimensionen statt 384 (die Werte sind frei erfunden)
hund  = np.array([0.70, -0.20, 0.50, 0.90])
katze = np.array([0.55, -0.10, 0.65, 0.90])
auto  = np.array([0.10, 0.70, -0.20, 0.30])

# Kosinus-Ähnlichkeit: beide Vektoren auf Länge 1 bringen, dann Skalarprodukt
# (setzt Vektoren ungleich dem Nullvektor voraus, sonst Division durch null)
def cos_sim(a, b):
    return (a / np.linalg.norm(a)) @ (b / np.linalg.norm(b))

print(f"cos(Hund, Katze) = {cos_sim(hund, katze):.2f}")
print(f"cos(Hund, Auto)  = {cos_sim(hund, auto):.2f}")
````

<div style="background:#0E2841; color:#fff; padding:6px 16px; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Kosinus-Ähnlichkeit</div>
<div style="background:#E9EDF4; color:#1a1a1a; padding:12px 16px; margin:0;">
<p>Gegeben seien zwei Vektoren $A$ und $B$, beide ungleich dem Nullvektor (sonst wäre der Nenner null). Ihre Kosinus-Ähnlichkeit ist der Kosinus des Winkels $\theta$ zwischen ihnen:</p>
<p>$$\cos(\theta) = \frac{A \cdot B}{|A| \, |B|}$$</p>
<details style="margin-top:10px"><summary style="cursor:pointer; color:#0E2841; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #0E284155"><p>Der Zähler ist das Skalarprodukt, ausgeschrieben $A \cdot B = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$: Jede Dimension trägt das Produkt der beiden Werte bei. Haben beide Vektoren in denselben Dimensionen hohe Werte, werden diese Beiträge groß, und die Summe auch. Der Nenner $|A| \, |B|$ teilt durch die Längen der Vektoren (die euklidischen Normen) und macht das Ergebnis damit von der Länge unabhängig. Gemessen wird so der Winkel zweier Richtungen: Der Wert 1 heißt gleiche Richtung, 0 senkrecht zueinander, −1 entgegengesetzt — das sind zunächst rein geometrische Aussagen. In Bedeutung übersetzt: Ein hoher Wert heißt, das Modell hält die Texte für sehr ähnlich; ein Wert nahe 0 oder darunter heißt nur, dass das Modell wenig Verbindung sieht. Sprachliche Gegensätze liegen dabei nicht bei −1: „heiß" und „kalt" tauchen in denselben Kontexten auf und bekommen darum sogar recht ähnliche Vektoren. Mathematisch liegt der Wert immer zwischen −1 und 1. Auch negative Werte kommen vor — in unseren Kursdaten bei gut einem Viertel der Frage-Absatz-Vergleiche, wenn auch nie unter −0,23; wie häufig und wie tief, hängt vom Modell und den verglichenen Texten ab.</p>
<p>Bringt man beide Vektoren vorher auf Länge 1, wird der Nenner zu 1, und vom Bruch bleibt nur das Skalarprodukt übrig. Genau deshalb besteht unsere Suche aus <strong>Normalize</strong> und dann <strong>Compare</strong>: Das ist diese Formel, arbeitsteilig aufgeschrieben.</p></div></details>
</div>
</div>

Warum ergibt ein Winkel von 0 genau den Wert 1? Ein Blick auf die Kosinus-Kurve zeigt es: Bei 0° startet sie bei 1, fällt bis 90° auf 0 und erreicht bei 180° den Wert −1:

````python
# Die Kosinus-Kurve: von Winkel 0 (gleiche Richtung) bis 180 Grad (entgegengesetzt)
import matplotlib.pyplot as plt

winkel = np.linspace(0, np.pi, 200)
fig, ax = plt.subplots(figsize=(5.5, 2.6))
ax.plot(np.degrees(winkel), np.cos(winkel), color="#156082", lw=2)
for grad in [0, 90, 180]:
    ax.scatter([grad], [np.cos(np.radians(grad))], color="#E97132", zorder=3)
ax.axhline(0, color="#adb5bd", lw=0.8)
ax.set_xticks([0, 45, 90, 135, 180])
ax.set_xlabel("Winkel zwischen den Vektoren in Grad")
ax.set_ylabel("Kosinus")
ax.set_title("Winkel 0 = Kosinus 1")
plt.show()
````

### <a id="kap-eine-rechnung-sieben-ebenen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.1.3&nbsp;&nbsp;Eine Rechnung, sieben Ebenen</span>

Bei 400 Absätzen ist die Suche in jedem Fall schnell. Echte Systeme durchsuchen aber Millionen von Dokumenten, und dann wird ein Schritt teuer: das Normalisieren und Vergleichen riesiger Zahlenmengen. Die Mathematik dahinter ist simpel: Jeder Vektor, also jede Zeile der Matrix, wird durch seine Länge geteilt. Aber ein Computer kann genau diese Rechnung auf sehr unterschiedliche Weise ausführen, und die Unterschiede sind gewaltig. Das schauen wir uns jetzt an.

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Sieben Ebenen, von der handgeschriebenen Python-Schleife bis zum eigenen GPU-Kernel. Was sie unterscheidet, ist vor allem, <strong>wer die Schleife ausführt</strong>: der Python-Interpreter Zeile für Zeile, eine vorkompilierte Bibliothek wie NumPy in Maschinencode, ein Compiler, der den Code erst zur Laufzeit übersetzt, oder Tausende GPU-Threads gleichzeitig. Die Rechnung selbst bleibt dabei identisch, nur der Weg vom Python-Text zur ausgeführten Arithmetik ändert sich — und zwischen der obersten und der untersten Ebene liegt ein Faktor im Tausenderbereich, den Sie gleich an echten Messwerten sehen. Jede Ebene ist eine eigene Station dieses Semesters. Für heute genügt eine grobe Orientierung: die Ebenen wiedererkennen und die Größenordnungen einordnen können.</p></div></details>

Damit der Vergleich fair ist, messen alle Ebenen dieselbe Aufgabe auf denselben Daten: einem **synthetischen Messsatz** aus 10.000 Zufallsvektoren mit je 384 Zahlen, aufgebaut wie unser Korpus, nur 25-mal größer (insgesamt 3,84 Millionen Zahlen). Synthetisch deshalb, weil Zufallsdaten mit festem Seed jederzeit exakt reproduzierbar sind und wir die Größe frei wählen können.

Gemessen wird mit `time.perf_counter()`, der präzisen Stoppuhr von Python: Uhr starten, Funktion ausführen, Uhr stoppen. Weil auf einem Rechner immer auch anderes läuft (Betriebssystem, Browser, Hintergrunddienste), messen wir mehrfach und nehmen den **Bestwert**: die schnellste beobachtete Ausführung unter den gegebenen Bedingungen. Er zeigt, was die Funktion kann, wenn wenig dazwischenfunkt; eine typische Laufzeit im Alltag liegt eher etwas darüber. Genau das tut die Funktion `measure`.

Noch ein Hinweis zu den Zahlen, die im Skript unter einigen Zellen abgedruckt sind: Sie sind **exemplarisch**, gemessen auf einem Referenzrechner. Führen Sie die Zellen selbst aus; Ihre Werte werden abweichen, auf die Größenordnungen kommt es an.

````python
import time

rng = np.random.default_rng(42)                             # fester Seed -> reproduzierbar
X = rng.standard_normal((10_000, 384)).astype(np.float32)   # der Messsatz: [10000, 384]

def measure(f, *args, repeats=3):
    # f mehrfach ausführen und die jeweils gestoppte Zeit sammeln
    times = []
    for _ in range(repeats):
        start = time.perf_counter()
        f(*args)
        times.append(time.perf_counter() - start)
    return min(times)                                       # Bestwert, siehe Text oben
````

**Ebene 1** ist der naive Weg: Python-Schleifen, wie man sie im ersten Semester schreibt. Jede einzelne der 3,84 Millionen Zahlen wird als Python-Objekt angefasst, Zeile für Zeile die Länge berechnet, dann jede Zahl durch diese Länge geteilt. Das funktioniert, aber Sie werden die Wartezeit spüren (deshalb messen wir hier nur einmal):

````python
# Ebene 1 — reines Python: jede Zahl einzeln anfassen
def normalize_python(X):
    result = []
    for row in X:
        length = sum(x * x for x in row) ** 0.5      # Länge der Zeile
        result.append([x / length for x in row])     # jede Zahl durch die Länge teilen
    return result

t_python = measure(normalize_python, X, repeats=1)
print(f"{t_python:.3f} s")
````

**Ebene 2** rechnet exakt dasselbe, aber als eine einzige NumPy-Zeile. Der Trick: Die Schleifen laufen jetzt nicht mehr im Python-Interpreter, sondern in vorkompiliertem Maschinencode über den ganzen Datenblock auf einmal. Warum das so viel ausmacht, ist das Thema der NumPy-Lerneinheit:

````python
# Ebene 2 — NumPy: dieselbe Rechnung, vektorisiert
def normalize_numpy(X):
    return X / np.linalg.norm(X, axis=1, keepdims=True)

t_numpy = measure(normalize_numpy, X)
print(f"{t_numpy * 1000:.2f} ms — Faktor {t_python / t_numpy:,.0f} gegenüber Ebene 1")
````

Das gilt für eine Colab-GPU-Laufzeit genauso wie für Ihren eigenen Rechner mit NVIDIA-Grafikkarte. Ohne GPU überspringt sie die Messung; die Tabelle unten nutzt dann den vorab gemessenen Wert oder verweist auf die Lerneinheit, in der diese Ebene drankommt. Eine Messentscheidung gehört dazugesagt: Gemessen wird mit Daten, die bereits im GPU-Speicher liegen — das Kopieren dorthin (`Xt.cuda()`) zählt nicht mit:

````python
# Ebene 4 — PyTorch auf der GPU (misst nur, wenn eine CUDA-GPU verfügbar ist)
t_gpu = None
if t_torch is not None and torch.cuda.is_available():
    Xg = Xt.cuda()                        # Messsatz in den GPU-Speicher kopieren
    def normalize_gpu():
        Xg / Xg.norm(dim=1, keepdim=True)
        torch.cuda.synchronize()          # auf die GPU warten — warum, zeigt das nächste Kapitel
    normalize_gpu()                       # einmal aufwärmen (auch das erklärt das nächste Kapitel)
    t_gpu = measure(normalize_gpu, repeats=10)
    print(f"{t_gpu * 1000:.2f} ms auf {torch.cuda.get_device_name(0)}")
else:
    print("Keine CUDA-GPU in dieser Umgebung — die Tabelle nutzt den vorab gemessenen Wert.")
````

### <a id="kap-messen-ist-schwerer-als-es-aussieht"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.1.4&nbsp;&nbsp;Messen ist schwerer, als es aussieht</span>

Laufzeiten sind die zentrale Währung dieses Moduls, und zugleich erstaunlich leicht falsch zu messen. Zwei klassische Fallen sollen Sie schon heute gesehen haben.

**Falle 1: Eine einzelne Messung ist keine verlässliche Zahl.** Auf jedem Rechner läuft nebenher anderes (Betriebssystem, Browser, Hintergrunddienste), und beim allerersten Aufruf kommen Einmal-Kosten dazu, etwa das Füllen der Caches. Eine Einzelmessung kann deshalb deutlich zu hoch ausfallen. Vergleichen Sie selbst, was eine Einzelmessung gegenüber dem Bestwert aus 20 Wiederholungen meldet (den Erstaufruf-Effekt zeigt diese Zelle nicht mehr, denn `normalize_numpy` ist oben schon gelaufen — den sähen Sie nur in einer frischen Sitzung):

````python
# Falle 1: Einzelmessung gegen Bestwert aus 20 Wiederholungen
single = measure(normalize_numpy, X, repeats=1)
best_of = measure(normalize_numpy, X, repeats=20)
print(f"einzelne Messung:   {single * 1000:.2f} ms")
print(f"Bestwert aus 20:    {best_of * 1000:.2f} ms")
````

**Falle 2: GPUs arbeiten asynchron.** Ein GPU-Aufruf kehrt sofort zurück: Gerechnet wird im Hintergrund weiter, während Python schon die nächste Zeile ausführt. Wer nur den Aufruf stoppt, misst also das Abschicken der Arbeit, nicht die Arbeit selbst. Bei unserem kleinen Normalize fällt das kaum auf, weil die Rechnung selbst nur Mikrosekunden dauert. Deshalb nehmen wir für die Gegenüberstellung eine deutlich größere Rechnung, eine Matrixmultiplikation. Mit einer CUDA-GPU messen Sie live, ohne kommt das Ergebnis aus der Messdatei:

````python
# Falle 2: naive GPU-Messung (ohne Warten) gegen korrekte Messung (mit Synchronisation)
# Eine große Matrixmultiplikation, damit der Unterschied deutlich sichtbar wird
if t_gpu is not None:
    _ = Xg @ Xg.T                         # einmal aufwärmen, damit beide Messungen
    torch.cuda.synchronize()              # aus demselben Zustand starten (siehe Falle 1)
    start = time.perf_counter()
    Xg @ Xg.T                             # ohne Warten: stoppt nur das Abschicken
    naive_ms = (time.perf_counter() - start) * 1000
    torch.cuda.synchronize()
    start = time.perf_counter()
    Xg @ Xg.T
    torch.cuda.synchronize()              # mit Warten: die echte Rechenzeit
    correct_ms = (time.perf_counter() - start) * 1000
    print(f"GPU, naiv gemessen:     {naive_ms:.2f} ms   (ohne Synchronisation)")
    print(f"GPU, korrekt gemessen:  {correct_ms:.2f} ms")
else:
    trap = timings["gpu_timing_trap"]
    if trap["naive_ms"] is not None:
        print(f"GPU, naiv gemessen:     {trap['naive_ms']:.2f} ms   (ohne Synchronisation)")
        print(f"GPU, korrekt gemessen:  {trap['correct_ms']:.2f} ms")
    else:
        print("Die GPU-Gegenüberstellung wird beim ersten Lauf auf der Kurs-Hardware gemessen.")
````

Wie man auf der GPU korrekt misst, mit Aufwärmen und Synchronisation, lernen Sie in der Lerneinheit zu Tensoren und Messung. Bis dahin gilt die Grundregel dieses Moduls: keiner Laufzeitzahl glauben, deren Messweg Sie nicht kennen.

## <a id="kap-python-und-sein-oekosystem"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:3px solid #156082; padding-bottom:4px;  padding-top:26px;"><span style="float:right"><span class="modulchip" style="font-size:0.55em; font-weight:bold; letter-spacing:0.02em; border:1.5px solid #475569; color:#475569; padding:1px 8px; border-radius:10px; margin-left:12px; vertical-align:0.35em; white-space:nowrap;">M1 · Python-Grundlagen</span></span>1.3&nbsp;&nbsp;Python und sein Ökosystem</span>

Bevor Sie im Lab die ersten Funktionen schreiben, ein kurzer Blick auf die Sprache selbst: woher sie kommt, wie sie ausgeführt wird und warum ausgerechnet sie die KI-Welt dominiert.

**Woher Python kommt.** Entworfen hat die Sprache Guido van Rossum, die erste Version erschien 1991. Der Name stammt von der Komikertruppe Monty Python, nicht von der Schlange. Das Entwurfsziel war von Anfang an gut lesbarer Code. Heute pflegt die Python Software Foundation die Sprache, eine neue Version erscheint jährlich.

**Python 2 und Python 3.** Lange gab es zwei Welten: 2008 erschien Python 3 und räumte an Stellen auf, die mit Python 2 unvereinbar waren (unter anderem Unicode-Strings und `print` als Funktion). Der Umstieg zog sich über ein Jahrzehnt; seit 2020 wird Python 2 nicht mehr gepflegt. Alles Aktuelle, auch dieser Kurs, ist Python 3. Alte Python-2-Beispiele im Netz erkennen Sie meist am `print` ohne Klammern.

**Philosophie.** Die Leitsätze der Sprache stehen im „Zen of Python" (geben Sie einmal `import this` in eine Codezelle ein): Lesbarkeit zählt, explizit ist besser als implizit, und es sollte einen offensichtlichen Weg geben, etwas zu tun. Daher kommt die Einrückung als Syntax, und daher kommt der kompakte Stil, den Sie im Lab sehen werden. Code, der diese Ideen nutzt, heißt „pythonisch".

**Wie Python ausgeführt wird.** Python ist eine interpretierte Sprache: Ein Programm, der **Interpreter**, liest Ihren Code und führt ihn Anweisung für Anweisung aus. Einen separaten Übersetzungsschritt, den Sie wie bei C++ selbst anstoßen, gibt es nicht; intern übersetzt der Interpreter den Quelltext automatisch in einen Zwischencode (Bytecode) und führt diesen aus. Die Standard-Implementierung dieses Interpreters heißt **CPython**, sie ist selbst in C geschrieben. Dieser Komfort kostet Geschwindigkeit, und genau das haben Sie in Ebene 1 des Appetizers gemessen. Einige dieser Abläufe unter der Haube schauen wir uns im Laufe des Semesters genauer an.

**Pakete und Ökosystem.** Die eigentliche Stärke der Sprache liegt neben ihr: Der Python Package Index (PyPI) hält Hunderttausende fertige Pakete bereit. Installiert werden sie mit einem **Package Manager** (`pip`, oder moderner `uv`, das auch unser Startskript nutzt), üblicherweise in eine **virtuelle Umgebung** je Projekt, damit sich die Abhängigkeiten verschiedener Projekte nicht in die Quere kommen. In Colab und auf dem Hochschul-Server sind die Kurspakete bereits installiert. Für Daten und KI ist dieses Ökosystem konkurrenzlos: NumPy, pandas, Matplotlib, scikit-learn und PyTorch sind allesamt Python-Pakete, und sie decken alle Stationen unserer Landkarte ab.

**Das Zwei-Sprachen-Muster.** Wie passt beides zusammen, ein vergleichsweise langsamer Interpreter und ein Ökosystem, das die rechenintensivste Disziplin der Informatik dominiert? Die Antwort ist ein Muster, das fast alle großen Pakete nutzen: Die Rechenkerne sind in einer kompilierten Sprache geschrieben, und Python dient als komfortable Bedienschicht darüber. NumPy rechnet in C, PyTorch in C++ und auf der GPU in CUDA, SciPy greift für lineare Algebra auf jahrzehntelang optimierte Fortran-Bibliotheken zurück, und die Tokenizer der Hugging-Face-Bibliotheken sind in Rust geschrieben. Ein Aufruf wie `np.dot(a, b)` verlässt deshalb den Interpreter und läuft als kompilierter Maschinencode; nur das Ergebnis kommt als Python-Objekt zurück. Sie schreiben also Python, gerechnet wird woanders. In der Fachliteratur heißt diese Arbeitsteilung das **Two-Language Problem**: Wer einen schnellen Rechenkern schreiben will, muss beide Welten beherrschen. Neuere Sprachen wie Julia oder Mojo setzen genau hier an und versprechen, Komfort und Geschwindigkeit in einer Sprache zu vereinen. Für dieses Muster spricht noch ein zweiter Grund: Innerhalb eines Prozesses führt CPython wegen des **Global Interpreter Lock** (GIL) immer nur einen Python-Thread zur Zeit aus (seit Python 3.14 gibt es daneben eine offiziell unterstützte Variante ohne GIL, die sich aber erst verbreiten muss). Parallel rechnen lässt sich trotzdem — mit mehreren Prozessen, oder eben in den kompilierten Rechenkernen, die den GIL während der Arbeit freigeben.

**Wo Python überall steckt.** Datenanalyse und KI sind das dominierende Einsatzgebiet, dazu kommen Web-Backends (Django, FastAPI), Automatisierung und Skripte, wissenschaftliches Rechnen und die Lehre. In Umfragen unter Entwicklerinnen und Entwicklern steht Python seit Jahren auf den vordersten Plätzen.

Wie die Ähnlichkeitswerte berechnet werden, kommt früh im Semester dran. Heute sind sie einfach da.

````python
# Kursdaten bereitstellen — nutzt data/, lädt Fehlendes aus dem Kursrepo nach
from pathlib import Path
from urllib.request import urlretrieve
for _name in ["appetizer_passages.json", "appetizer_embeddings.npy", "appetizer_queries.json", "appetizer_query_embeddings.npy"]:
    _ziel = Path("data") / _name
    if not _ziel.exists():
        _ziel.parent.mkdir(parents=True, exist_ok=True)
        urlretrieve("https://raw.githubusercontent.com/skuhswo/tensor-to-gpu-material/main/data/" + _name, _ziel)
    print(_ziel, "\u2713")
````

````python
# Vorgabe: Daten und Trefferlisten bereitstellen — einfach ausführen
import json
import numpy as np

corpus = json.load(open("data/appetizer_passages.json"))
passages = corpus["passages"]
E = np.load("data/appetizer_embeddings.npy")
queries = json.load(open("data/appetizer_queries.json"))["queries"]
Q = np.load("data/appetizer_query_embeddings.npy")

def get_hits(query_no):
    # liefert [(title, score), …] für die Beispielanfrage query_no — unsortiert
    q = Q[query_no] / np.linalg.norm(Q[query_no])
    scores = (E / np.linalg.norm(E, axis=1, keepdims=True)) @ q
    return [(passages[i]["title"], float(scores[i])) for i in range(len(passages))]

hits = get_hits(0)
print("Anfrage:", queries[0]["text"])
print(len(hits), "Treffer, die ersten drei (unsortiert):", hits[:3])
````

Modernes C++ kennt beides übrigens auch (`for (const auto& [title, score] : hits)`), in Python ist es nur der Normalfall statt der Ausnahme:
