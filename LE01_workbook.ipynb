### <a id="kap-die-suche-ausprobieren"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.2.1&nbsp;&nbsp;Die Suche ausprobieren</span>

Unsere Suchmaschine besteht aus zwei Zutaten, die im Ordner `data/` liegen und die die nächste Zelle bei Bedarf automatisch nachlädt:

- **400 Absätze aus der deutschen Wikipedia** — das ist unser Dokumentbestand, der „Korpus". Jeder Absatz stammt aus einem anderen Artikel.
- **Je Absatz ein Vektor aus 384 Zahlen** — die sogenannten Embeddings. Ein Sprachmodell hat jeden Absatz vorab in einen solchen Zahlenvektor übersetzt; wie, sehen wir später im Semester.

Dazu kommen acht **Beispielanfragen** (echte Fragen zu diesem Korpus), ebenfalls schon als Text und als Vektor. So funktioniert die Suche auch ohne installiertes Sprachmodell.

````python
# Kursdaten bereitstellen — nutzt data/, lädt Fehlendes aus dem Kursrepo nach
from pathlib import Path
from urllib.request import urlretrieve
for _name in ["appetizer_passages.json", "appetizer_embeddings.npy", "appetizer_queries.json", "appetizer_query_embeddings.npy", "appetizer_timings.json"]:
    _ziel = Path("data") / _name
    if not _ziel.exists():
        _ziel.parent.mkdir(parents=True, exist_ok=True)
        urlretrieve("https://raw.githubusercontent.com/skuhswo/tensor-to-gpu-material/main/data/" + _name, _ziel)
    print(_ziel, "\u2713")
````

````python
import json
import numpy as np

# Korpus laden: die Absätze als Text und die zugehörigen Vektoren als Matrix.
# E hat die Form [400, 384]: eine Zeile je Absatz, 384 Zahlen je Zeile.
corpus = json.load(open("data/appetizer_passages.json"))
passages = corpus["passages"]
E = np.load("data/appetizer_embeddings.npy")

# Die Beispielanfragen: ebenfalls Text plus je ein Vektor (Matrix Q, Form [8, 384]).
queries = json.load(open("data/appetizer_queries.json"))["queries"]
Q = np.load("data/appetizer_query_embeddings.npy")

print(len(passages), "Passagen ·", E.shape[1], "Zahlen je Vektor ·", len(queries), "Beispielanfragen")
````

Werfen wir zuerst einen Blick in die Daten. So sehen drei der 400 Absätze aus — jeder hat eine Nummer (`id`), einen Artikeltitel und den eigentlichen Text:

````python
# Drei Beispiel-Absätze anzeigen (von 400)
for p in passages[:3]:
    print(f"[{p['id']}] {p['title']}")
    print("    " + p["text"][:150].replace("\n", " ") + " …\n")
````

Und das sind die acht Beispielanfragen. Die Nummer vorne brauchen Sie gleich, um eine Anfrage auszuwählen:

````python
# Alle Beispielanfragen mit ihrer Nummer
for no, q in enumerate(queries):
    print(f"[{no}] {q['text']}")
````

Texte mit ähnlicher Bedeutung bekommen vom Sprachmodell ähnliche Vektoren — „Wo liegt San Diego?" und ein Absatz über die Geographie San Diegos zeigen im 384-dimensionalen Raum in fast dieselbe Richtung. Suchen heißt deshalb nur noch: den Vektor der Frage mit allen 400 Absatz-Vektoren vergleichen und die ähnlichsten heraussuchen. Drei Schritte: **Normalize** (alle Vektoren auf Länge 1 bringen, damit nur die Richtung zählt, nicht die Länge), **Compare** (Skalarprodukt der Frage mit jedem Absatz — bei Länge 1 ist das genau die Kosinus-Ähnlichkeit: 1 = gleiche Richtung, 0 = nichts gemein), **Select** (die k besten Treffer nehmen).

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Genau diese drei Schritte begleiten Sie als Semesterfall durch alle Lerneinheiten: erst in NumPy, später in PyTorch, am Ende als eigener GPU-Kernel. Warum das Skalarprodukt normierter Vektoren die Kosinus-Ähnlichkeit ist und wie ein Sprachmodell zu seinen Vektoren kommt, klären wir dort Schritt für Schritt.</p></div></details>

Die folgende Zelle setzt diese drei Schritte in wenigen Zeilen um. Die Funktion `search(query_no, k)` hat zwei Parameter: `query_no` ist die **Nummer der Beispielanfrage** aus der Liste oben (0 bis 7), und `k` bestimmt, **wie viele Treffer** angezeigt werden. Hinter jedem Treffer steht sein Ähnlichkeitswert — je näher an 1, desto besser passt der Absatz zur Frage:

````python
def normalize(v):
    # Jede Zeile durch ihre Länge teilen -> alle Vektoren haben Länge 1
    return v / np.linalg.norm(v, axis=-1, keepdims=True)

def search(query_no, k=3):
    q = normalize(Q[query_no])               # Normalize: der Fragevektor
    scores = normalize(E) @ q                # Compare: Kosinus-Ähnlichkeit zu allen 400 Absätzen
    best = np.argsort(scores)[::-1][:k]      # Select: die k höchsten Werte
    print("Anfrage:", queries[query_no]["text"], "\n")
    for pid in best:
        p = passages[pid]
        print(f"  {scores[pid]:.3f}  {p['title']}")
        print("         " + p["text"][:160].replace("\n", " ") + " …\n")

# Anfrage Nr. 0 ("Wo liegt San Diego?"), drei Treffer:
search(0)
````

````python
# Probieren Sie es selbst: andere Anfrage-Nummer (0 bis 7), mehr Treffer über k.
# Hier: Anfrage Nr. 3 mit den fünf besten Treffern.
search(3, k=5)
````

Achten Sie beim Ausprobieren auf zwei Dinge: Der beste Treffer hat meist einen deutlich höheren Ähnlichkeitswert als der Rest — die Suche ist sich sicher. Und die Treffer teilen oft kein einziges Wort mit der Frage; verglichen wird Bedeutung, nicht Buchstaben.

### <a id="kap-eine-rechnung-sieben-ebenen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.2.2&nbsp;&nbsp;Eine Rechnung, sieben Ebenen</span>

Bei 400 Absätzen ist die Suche in jedem Fall schnell. Echte Systeme durchsuchen aber Millionen von Dokumenten, und dann wird ein Schritt teuer: das Normalisieren und Vergleichen riesiger Zahlenmengen. Die Mathematik dahinter ist simpel — jede Zeile der Matrix durch ihre Länge teilen. Aber ein Computer kann genau diese Rechnung auf sehr unterschiedliche Weise ausführen, und die Unterschiede sind gewaltig. Das schauen wir uns jetzt an.

<details style="margin-top:10px"><summary style="cursor:pointer; color:#1F5FB0; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #1F5FB055"><p>Sieben Ebenen, von der handgeschriebenen Python-Schleife bis zum eigenen GPU-Kernel — jede Ebene ist eine eigene Station dieses Semesters.</p></div></details>

Damit der Vergleich fair ist, messen alle Ebenen dieselbe Aufgabe auf denselben Daten: einem **synthetischen Messsatz** aus 10.000 Zufallsvektoren mit je 384 Zahlen — wie unser Korpus, nur 25-mal größer (insgesamt 3,84 Millionen Zahlen). Synthetisch deshalb, weil Zufallsdaten mit festem Seed jederzeit exakt reproduzierbar sind und wir die Größe frei wählen können.

Gemessen wird mit `time.perf_counter()`, der präzisen Stoppuhr von Python: Uhr starten, Funktion ausführen, Uhr stoppen. Weil auf einem Rechner immer auch anderes läuft (Betriebssystem, Browser, Hintergrunddienste), messen wir mehrfach und nehmen den **Bestwert** — der kommt der wahren Rechenzeit am nächsten, alles darüber ist Störung. Genau das tut die Funktion `measure`.

Noch ein Hinweis zu den Zahlen, die im Skript unter einigen Zellen abgedruckt sind: Sie sind **exemplarisch** — gemessen auf einem Referenzrechner. Führen Sie die Zellen selbst aus; Ihre Werte werden abweichen, auf die Größenordnungen kommt es an.

````python
import time

rng = np.random.default_rng(42)                             # fester Seed -> reproduzierbar
X = rng.standard_normal((10_000, 384)).astype(np.float32)   # der Messsatz: [10000, 384]

def measure(f, *args, repeats=3):
    # f mehrfach ausführen und die jeweils gestoppte Zeit sammeln
    times = []
    for _ in range(repeats):
        start = time.perf_counter()
        f(*args)
        times.append(time.perf_counter() - start)
    return min(times)                                       # Bestwert, siehe Text oben
````

**Ebene 1** ist der naive Weg: Python-Schleifen, wie man sie im ersten Semester schreibt. Jede einzelne der 3,84 Millionen Zahlen wird als Python-Objekt angefasst — Zeile für Zeile die Länge berechnen, dann jede Zahl durch diese Länge teilen. Das funktioniert, aber Sie werden die Wartezeit spüren (deshalb messen wir hier nur einmal):

````python
# Ebene 1 — reines Python: jede Zahl einzeln anfassen
def normalize_python(X):
    result = []
    for row in X:
        length = sum(x * x for x in row) ** 0.5      # Länge der Zeile
        result.append([x / length for x in row])     # jede Zahl durch die Länge teilen
    return result

t_python = measure(normalize_python, X, repeats=1)
print(f"{t_python:.3f} s")
````

**Ebene 2** rechnet exakt dasselbe, aber als eine einzige NumPy-Zeile. Der Trick: Die Schleifen laufen jetzt nicht mehr im Python-Interpreter, sondern in vorkompiliertem Maschinencode über den ganzen Datenblock auf einmal. Warum das so viel ausmacht, ist das Thema der NumPy-Lerneinheit:

````python
# Ebene 2 — NumPy: dieselbe Rechnung, vektorisiert
def normalize_numpy(X):
    return X / np.linalg.norm(X, axis=1, keepdims=True)

t_numpy = measure(normalize_numpy, X)
print(f"{t_numpy * 1000:.2f} ms — Faktor {t_python / t_numpy:,.0f} gegenüber Ebene 1")
````

**Ebene 3** ist dieselbe Rechnung in PyTorch, dem Framework, in dem heute die meisten neuronalen Netze trainiert werden — noch auf der CPU, aber schon mit dem Werkzeug, das später nahtlos auf die GPU wechselt:

````python
# Ebene 3 — PyTorch auf der CPU (in Colab vorinstalliert)
try:
    import torch
    Xt = torch.from_numpy(X)
    t_torch = measure(lambda: Xt / Xt.norm(dim=1, keepdim=True))
    print(f"{t_torch * 1000:.2f} ms")
except ImportError:
    t_torch = None
    print("PyTorch ist in dieser Umgebung nicht installiert — diese Ebene holen wir in Colab nach.")
````

**Ebene 4** ist derselbe PyTorch-Aufruf, nur auf der GPU. Diese Zelle misst live, wenn in Ihrer Umgebung eine CUDA-GPU steckt — das gilt für eine Colab-GPU-Laufzeit genauso wie für Ihren eigenen Rechner mit NVIDIA-Grafikkarte. Ohne GPU überspringt sie die Messung; die Tabelle unten nutzt dann den vorab gemessenen Wert oder verweist auf die Lerneinheit, in der diese Ebene drankommt:

````python
# Ebene 4 — PyTorch auf der GPU (misst nur, wenn eine CUDA-GPU verfügbar ist)
t_gpu = None
if t_torch is not None and torch.cuda.is_available():
    Xg = Xt.cuda()                        # Messsatz in den GPU-Speicher kopieren
    def normalize_gpu():
        Xg / Xg.norm(dim=1, keepdim=True)
        torch.cuda.synchronize()          # auf die GPU warten — warum, zeigt das nächste Kapitel
    normalize_gpu()                       # einmal aufwärmen (auch das erklärt das nächste Kapitel)
    t_gpu = measure(normalize_gpu, repeats=10)
    print(f"{t_gpu * 1000:.2f} ms auf {torch.cuda.get_device_name(0)}")
else:
    print("Keine CUDA-GPU in dieser Umgebung — die Tabelle nutzt den vorab gemessenen Wert.")
````

Die **Ebenen 5 bis 7** — der PyTorch-Compiler, ein handgeschriebener CUDA-Kernel und die GPU-Sprache Mojo — brauchen Werkzeuge, die Sie erst später im Semester haben. Die nächste Zelle setzt alles zu einer Tabelle zusammen: Wo schon eine Zeit vorliegt (live gemessen oder vorab auf der Kurs-GPU), steht sie mit ihrem Faktor gegenüber der Python-Schleife von Ebene 1. Wo noch keine steht, zeigt die Tabelle stattdessen die Lerneinheit, in der Sie diese Ebene selbst bauen und messen werden:

````python
# Live-Messungen und gespeicherte GPU-Messungen zu einer Tabelle zusammensetzen
timings = json.load(open("data/appetizer_timings.json"))

def print_row(name, sec, le=None):
    if sec is not None:
        print(f"{name:28s}{sec * 1000:>9.2f} ms{t_python / sec:>9,.0f}×")
    else:
        info = f"kommt in Lerneinheit {le}" if le else "—"
        print(f"{name:28s}{info:>25s}")

print(f"{'Ebene':28s}{'Zeit':>12s}{'Faktor':>10s}")
print_row("1  Python-Schleife", t_python)
print_row("2  NumPy", t_numpy)
print_row("3  PyTorch (CPU)", t_torch, le=6)
for name, entry in timings["levels"].items():
    sec = entry["sec"]
    if name.startswith("4") and t_gpu is not None:
        sec = t_gpu                       # Live-Messung ersetzt den gespeicherten Wert
    print_row(name, sec, entry.get("le"))

quelle = torch.cuda.get_device_name(0) if t_gpu is not None else timings["device"]
if quelle:
    print("\nGPU-Werte gemessen auf:", quelle)
````

### <a id="kap-messen-ist-schwerer-als-es-aussieht"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.2.3&nbsp;&nbsp;Messen ist schwerer, als es aussieht</span>

Laufzeiten sind die zentrale Währung dieses Moduls — und zugleich erstaunlich leicht falsch zu messen. Zwei klassische Fallen sollen Sie schon heute gesehen haben.

**Falle 1: Der erste Aufruf ist nicht repräsentativ.** Beim ersten Mal muss der Rechner noch Daten in die Caches holen und Code vorbereiten; ab dem zweiten Aufruf ist alles warm. Vergleichen Sie selbst, was ein einzelner (kalter) Aufruf gegenüber dem Bestwert aus 20 Wiederholungen meldet:

````python
# Falle 1: einzelner (kalter) Aufruf gegen Bestwert aus 20 Wiederholungen
single = measure(normalize_numpy, X, repeats=1)
best_of = measure(normalize_numpy, X, repeats=20)
print(f"einzelner Aufruf:   {single * 1000:.2f} ms")
print(f"Bestwert aus 20:    {best_of * 1000:.2f} ms")
````

**Falle 2: GPUs arbeiten asynchron.** Ein GPU-Aufruf kehrt sofort zurück — gerechnet wird im Hintergrund weiter, während Python schon die nächste Zeile ausführt. Wer nur den Aufruf stoppt, misst also das Abschicken der Arbeit, nicht die Arbeit selbst. Mit einer CUDA-GPU messen Sie die Gegenüberstellung live; ohne kommt sie aus der Messdatei:

````python
# Falle 2: naive GPU-Messung (ohne Warten) gegen korrekte Messung (mit Synchronisation)
if t_gpu is not None:
    start = time.perf_counter()
    Xg / Xg.norm(dim=1, keepdim=True)     # ohne Warten: stoppt nur das Abschicken
    naive_ms = (time.perf_counter() - start) * 1000
    torch.cuda.synchronize()
    start = time.perf_counter()
    Xg / Xg.norm(dim=1, keepdim=True)
    torch.cuda.synchronize()              # mit Warten: die echte Rechenzeit
    correct_ms = (time.perf_counter() - start) * 1000
    print(f"GPU, naiv gemessen:     {naive_ms:.2f} ms   (ohne Synchronisation)")
    print(f"GPU, korrekt gemessen:  {correct_ms:.2f} ms")
else:
    trap = timings["gpu_timing_trap"]
    if trap["naive_ms"] is not None:
        print(f"GPU, naiv gemessen:     {trap['naive_ms']:.2f} ms   (ohne Synchronisation)")
        print(f"GPU, korrekt gemessen:  {trap['correct_ms']:.2f} ms")
    else:
        print("Die GPU-Gegenüberstellung wird beim ersten Lauf auf der Kurs-Hardware gemessen.")
````

Wie man auf der GPU korrekt misst — mit Aufwärmen und Synchronisation — lernen Sie in der Lerneinheit zu Tensoren und Messung. Bis dahin gilt die Grundregel dieses Moduls: keiner Laufzeitzahl glauben, deren Messweg Sie nicht kennen.

### <a id="kap-variablen-und-typen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.6.1&nbsp;&nbsp;Variablen und Typen</span>

In Python deklarieren Sie keine Typen — eine Zuweisung genügt, und der Typ hängt am Wert, nicht an der Variablen. Das fühlt sich nach C++ erst frei und dann verdächtig an; wie Python damit trotzdem Ordnung hält, sehen wir am Kapitelende.

````python
# Keine Deklaration, kein Semikolon — der Typ hängt am Wert
count = 3                 # int
distance = 2.5            # float
name = "Tensor to GPU"    # str
ready = True              # bool

print(type(count), type(distance), type(name), type(ready))
````

````python
# f-Strings: Werte direkt in Text einsetzen (das printf/cout-Äquivalent)
print(f"{name} startet mit {count} Einheiten, Distanz {distance:.1f}")
````

### <a id="kap-kontrollfluss-einrueckung-statt-klammern"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.6.2&nbsp;&nbsp;Kontrollfluss: Einrückung statt Klammern</span>

Python kennt keine geschweiften Klammern: **Die Einrückung ist die Blockstruktur.** Was in C++ Konvention ist, ist hier Syntax. Und die `for`-Schleife läuft nicht über Indizes, sondern direkt über die Elemente:

````python
# if/elif/else — Doppelpunkt und Einrückung statt { }
score = 0.83
if score >= 0.8:
    print("starker Treffer")
elif score >= 0.5:
    print("brauchbar")
else:
    print("vermutlich daneben")
````

````python
# for läuft über die Elemente selbst — kein for (int i = 0; ...)
scores = [0.83, 0.41, 0.77, 0.12]
for s in scores:
    print(s)

# Wenn Sie doch einen Zähler brauchen: range bzw. enumerate
for i, s in enumerate(scores):
    print(i, "->", s)
````

### <a id="kap-funktionen"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.6.3&nbsp;&nbsp;Funktionen</span>

Funktionen beginnen mit `def`, brauchen keine Typangaben und können Parametern Standardwerte geben. Mit einem Tupel geben sie auch mehrere Werte auf einmal zurück:

````python
def top_score(scores, threshold=0.5):
    # größten Wert und die Anzahl der Werte über der Schwelle zurückgeben
    best = max(scores)
    hits = 0
    for s in scores:
        if s >= threshold:
            hits += 1
    return best, hits

best, hits = top_score([0.83, 0.41, 0.77, 0.12])
print(f"bester Wert {best}, {hits} über der Schwelle")
print(top_score([0.83, 0.41, 0.77, 0.12], threshold=0.8))
````

### <a id="kap-variablen-sind-etiketten"></a><span class="zwischentitel" style="display:block; color:#0E2841; border-bottom:1px solid rgba(21,96,130,0.45); padding-bottom:2px;  padding-top:16px;">1.6.4&nbsp;&nbsp;Variablen sind Etiketten</span>

<div style="border-radius:6px; overflow:hidden; margin:2px 0; margin-right:-20px;">
<div style="background:#C62828; color:#fff; padding:6px 16px; font-weight:bold; font-size:1.08em; margin:0; overflow:hidden;">Definition: Referenzsemantik</div>
<div style="background:#FDECEA; color:#1a1a1a; padding:12px 16px; margin:0;">
<p>Eine Python-Variable ist ein <strong>Name, der auf ein Objekt zeigt</strong> — wie ein Etikett an einem Karton. Eine Zuweisung <code>b = a</code> kopiert nie das Objekt, sondern klebt nur ein zweites Etikett auf denselben Karton.</p>
<details style="margin-top:10px"><summary style="cursor:pointer; color:#C62828; font-weight:600; font-size:0.9em">Mehr dazu</summary><div style="margin-top:8px; padding-top:8px; border-top:1px dashed #C6282855"><p>In C++ ist <code>b = a</code> eine Kopie des Werts: <code>b</code> und <code>a</code> leben danach getrennt. In Python teilen sich <code>a</code> und <code>b</code> dasselbe Objekt — Änderungen über den einen Namen sieht auch der andere. Das ist der wichtigste Denk-Unterschied zwischen beiden Sprachen und die Quelle der häufigsten Umsteiger-Fehler.</p></div></details>
</div>
</div>

````python
# Zwei Etiketten, ein Karton
a = [1, 2]
b = a              # KEINE Kopie — b zeigt auf dieselbe Liste
b.append(3)
print("a =", a)    # a sieht die Änderung mit
print("b =", b)
print("dasselbe Objekt?", a is b)
````

````python
# Wer wirklich eine Kopie will, sagt es ausdrücklich
a = [1, 2]
b = list(a)        # neue Liste mit denselben Elementen
b.append(3)
print("a =", a, "  b =", b, "  dasselbe Objekt?", a is b)
````

Der Unterschied zwischen `is` und `==`: `==` fragt „gleicher Inhalt?", `is` fragt „dasselbe Objekt?". Und nicht alles lässt sich überhaupt ändern: Listen sind **veränderlich** (mutable), Zahlen, Strings und Tupel sind **unveränderlich** (immutable) — bei `x = x + 1` entsteht ein neues Objekt, der alte Karton bleibt unberührt. Warum das für NumPy-Arrays und später für GPU-Speicher wichtig wird, sehen Sie ab Lerneinheit 3.

````python
# Vorgabe: Trefferlisten bereitstellen — einfach ausführen
def get_hits(query_no):
    # liefert [(title, score), …] für die Beispielanfrage query_no — unsortiert
    q = Q[query_no] / np.linalg.norm(Q[query_no])
    scores = (E / np.linalg.norm(E, axis=1, keepdims=True)) @ q
    return [(passages[i]["title"], float(scores[i])) for i in range(len(passages))]

hits = get_hits(0)
print("Anfrage:", queries[0]["text"])
print(len(hits), "Treffer, die ersten drei (unsortiert):", hits[:3])
````
